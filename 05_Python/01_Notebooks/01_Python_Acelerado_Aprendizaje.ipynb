#### Tuplas

Se crean con paréntesis. 

Pueden tener cualquier tipo de elemento (números, cadenas, otras listas, ...) incluso mezclados. 

Son inmutables (tienen una longitud fija y no pueden cambiarse sus elementos).

````python
# Pregunta: en que se parece una tupla a una lista?
# Respuesta: tambien es una secuencia ordenada. La diferencia clave es que la tupla no se puede modificar.
tupla1 = (1,'rojo',['Isaac',41])
tupla1
````

````python
# Pregunta: por que esta linea falla?
# Respuesta: las tuplas son inmutables: no permiten cambiar sus elementos por posicion.
tupla1[0] = 2
````

````python
#También se puede crear con la función tuple sobre una lista
tupla1 = tuple([1,2,3])
tupla1
````

````python
# Pregunta: por que (5) no es una tupla?
# Respuesta: los parentesis solos agrupan una expresion. La coma es lo que crea la tupla de un elemento.
# Tupla de un elemento: (5,) a ver su utilidad
print(type((5)))
print(type((5,)))
````

````python
# Pregunta: por que 1,2,3,4 es una tupla sin parentesis?
# Respuesta: en Python, la coma crea la tupla; los parentesis solo ayudan a leerla.
sin_formato = 1,2,3,4
type(sin_formato)
````

#### Diccionarios

Los diccionarios se crean con llaves. 

Son pares no ordenados y mutables de clave-valor. 

Pueden contener cualquier tipo de información, incluso mezclada.

````python
# Pregunta: que es un diccionario?
# Respuesta: una estructura de pares clave-valor. Buscamos por la clave, no por posicion.
dicc1 = {'nombre':'Isaac', 'edad':46}
dicc1
````

````python
# Pregunta: por que dict recibe pares?
# Respuesta: cada tupla tiene una clave y su valor: ('nombre','Isaac').
dicc1 = dict([('nombre','Isaac'),('edad',46)])
dicc1
````

````python
#Crear un diccionario vacío
vacio = {}
vacio
````

````python
# Pregunta: como cambio un valor de un diccionario?
# Respuesta: accedo por su clave y asigno un valor nuevo.
dicc1['edad'] = 43
dicc1
````

````python
# Pregunta: del dicc1['edad'] borra que cosa?
# Respuesta: borra el par completo cuya clave es 'edad'.
del dicc1['edad']
dicc1
````

#### Conjuntos (sets)

Los conjuntos se crean también con llaves, pero no son pares. 

Son colecciones de elementos UNICOS. 

Pueden contener cualquier tipo de información, incluso mezclada. Son mutables.

````python
# Pregunta: por que desaparecen los repetidos?
# Respuesta: un set solo guarda valores unicos; no mantiene duplicados.
conj1 = {1,1,1,2,2,2,'rojo','rojo'}
conj1
````

````python
# Pregunta: para que sirve set(lista)?
# Respuesta: convierte la lista en conjunto y elimina duplicados.
lista = [1,1,1,2,2,2,'rojo','rojo']
conj2 = set(lista)
conj2
````

````python
conj2 = set(lista)
````

````python
conj2
````

````python
#Podemos añadir un nuevo elemento con add
conj2.add('amarillo')
conj2
````

````python
# Pregunta: update es como append?
# Respuesta: parecido, pero update anade varios elementos de otro iterable al conjunto.
conj3 = {'negro','rosa'}
conj2.update(conj3)
conj2
````

````python
#Eliminar por valor con discard
conj2.discard('rosa')
conj2
````

````python
# Pregunta: que es la union de conjuntos?
# Respuesta: junta todos los elementos de ambos conjuntos sin repetir.
conj1 = {'negro','rosa'}
conj2 = {'rosa','gris'}
conj1.union(conj2)
````

````python
# Pregunta: que es la interseccion?
# Respuesta: solo los elementos que aparecen en los dos conjuntos.
conj1 = {'negro','rosa'}
conj2 = {'rosa','gris'}
conj1.intersection(conj2)
````

````python
# Pregunta: que significa difference?
# Respuesta: elementos que estan en conj1 pero no en conj2.
conj1 = {'negro','rosa'}
conj2 = {'rosa','gris'}
conj1.difference(conj2)
````

````python
# Pregunta: por que cambia el resultado si invierto los conjuntos?
# Respuesta: la diferencia depende del orden: conj2 - conj1 no es lo mismo que conj1 - conj2.
conj1 = {'negro','rosa'}
conj2 = {'rosa','gris'}
conj2.difference(conj1)
````

## INDEXACIÓN

Lo más importante es recordar siempre que en Python los índices empiezan en cero

````python
# Pregunta: por que lista[0] devuelve el primer elemento y no lista[1]?
# Respuesta: Python empieza a contar desde 0 porque los indices representan desplazamientos desde el inicio.
# Desplazamiento 0 = primer elemento, desplazamiento 1 = segundo elemento, etc.
lista = [5,8,7,3]
lista[0]
````

````python
# Pregunta: por que lista[1:3] devuelve dos elementos?
# Respuesta: toma posiciones 1 y 2, pero no la posicion 3.
# Con lista = [5,8,7,3], el resultado sera [8, 7].
lista[1:3]
````

### Indexar listas

````python
lista = ['rojo','verde','azul']
lista[2]
````

````python
# Funciona slice tambien en listas.
# lista[0:2] toma desde la posicion 0 hasta antes de la 2: posiciones 0 y 1.
lista[0:2]
````

````python
# Pregunta: por que lista['rojo'] falla?
# Respuesta: las listas se indexan por posicion numerica, no por valor.
lista['rojo']
````

````python
# Pregunta: como encuentro la posicion de un valor?
# Respuesta: index(valor) devuelve la primera posicion donde aparece ese valor.
lista = ['rojo','verde','rojo']
posicion = lista.index('rojo')
lista[posicion]
````

````python
# El problema es que solo devuelve la primera aparicion. Si queremos todas las posiciones, necesitamos un bucle o una comprension de lista.
posicion
````

````python
# para tener todas las posiciones de un valor en una lista, podemos usar una comprensión de lista con enumerate().
lista2 = ['rojo', 'verde', 'rojo']

posiciones = [i for i, color in enumerate(lista2) if color == 'rojo']
posiciones
````

````python
#La indexación con números negativos lo que hace es empezar por el final
lista = ['rojo','verde','azul']
lista[-1]
````

````python
# Pregunta: como leo una lista dentro de otra lista?
# Respuesta: voy por pasos. Primero lista_anidada[2], luego dentro de esa lista [1].
lista_anidada = ['Isaac','Maria',['Jose Antonio','Jose Manuel'],'Pedro']
print(lista_anidada[2])
print(lista_anidada[2][1])
````

### If

Se utiliza para comprobar condiciones si ... entonces ... Detrás del si va un criterio, que si se se cumple se aplica el proceso del entonces.

Para mantener una sintaxis limpia Python utiliza (en esto y en todo) la indentación: cada línea se sabe a qué bloque pertenece por su nivel de indentación.

````python
# Pregunta: que hace if?
# Respuesta: ejecuta el bloque indentado solo si la condicion es True.
x = 5
y = 3

if x > y:
    print('x es mayor que y')
````

````python
# Pregunta: cuando entra el else?
# Respuesta: cuando la condicion del if no se cumple.
x = 3
y = 5
if x > y:
    print('x es mayor que y')
else:
    print('y es mayor que x')
````

Notar la importancia de la indentación del código en Python. Esto no funciona:

````python
# Pregunta: por que este codigo no funciona?
# Respuesta: else debe estar alineado con if. La indentacion indica a que bloque pertenece cada linea.
x = 3
y = 5
if x > y:
    print('x es mayor que y')
    else:
        print('y es mayor que x')
````

Se pueden incluir condiciones intermedias con elif

````python
# Pregunta: para que sirve elif?
# Respuesta: permite probar otra condicion si el if anterior no se cumple.
x = 3
y = 6
if x > y:
    print('x es mayor que y')
elif x == y:
    print('x es igual que y')
else:
    print('y es mayor que x')
````

### Contadores

````python
# Pregunta: por que el contador empieza en 1?
# Respuesta: porque queremos imprimir desde 1 hasta antes del limite. Podria empezar en 0 si ese fuera el objetivo.
limite = 5
cont = 1
while cont < limite:
        print(cont)
        cont = cont + 1
print('Hasta aqui he llegado')
````

````python
# Pregunta: que significa cont += 1?
# Respuesta: es una forma corta de escribir cont = cont + 1.
limite = 5
cont = 1
while cont < limite:
        print(cont)
        cont += 1
print('Hasta aqui he llegado')
````

### Funciones definidas por el usuario

Quizá sin saberlo, hemos estado usando un montón de funciones. Sin ir más lejos print es una. 

En general una función es algo a lo que le pasas unos parámetros (aunque no necesariamente) y te devuelve una salida (aunque no necesariamente).

Pero además de las funciones predefinidas de Python nosotros podemos crear nuestras propias funciones.

Especificamos lo que queremos que devuelva con return.

````python
# Pregunta: que partes tiene una funcion?
# Respuesta: def + nombre + parametros. return indica que valor devuelve la funcion.
def suma_dos(num1,num2):
    resultado = num1 + num2
    return(resultado)
````

````python
# Pregunta: como uso una funcion despues de definirla?
# Respuesta: llamandola por su nombre y pasando argumentos entre parentesis.
suma_dos(2,4)
````

````python
# Pregunta: que significa num2 = 4 en los parametros?
# Respuesta: es un valor por defecto. Si no pasamos num2, Python usa 4.
def suma_dos(num1,num2 = 4):
    resultado = num1 + num2
    return(resultado)
suma_dos(num1 = 2)
````

````python
# Pregunta: una funcion puede devolver mas de un resultado?
# Respuesta: si. Python los devuelve como una tupla, y podemos desempaquetarlos en variables.
def suma_dos(num1,num2 = 4):
    resultado1 = num1 + num2
    resultado2 = num1 * num2
    return(resultado1,resultado2)

res1,res2 = suma_dos(num1 = 2)
print(res1)
print(res2)
````

````python
res3,res4 = suma_dos(6) # Aqui usamos el valor por defecto num2 = 4, porque no pasamos un segundo argumento. python comprende que 5 es num 1 => num1 = 5, num2 = 4
print(res3)
print(res4)
````

````python
# Pregunta: que son num1: int y -> int?
# Respuesta: son pistas de tipo. Ayudan a leer el codigo, pero no convierten automaticamente los datos.
def suma_dos(num1: int, num2: int) -> int: 
    resultado = num1 + num2
    return(resultado)
suma_dos(2, 5)
````

También se llaman anónimas, ya que no tienen nombre.

### Filter

Devuelve los elementos de una secuencia que cumplen una condición. Pero devuelve un objeto filter, que normalmente hay que convertir para visualizarlo, por ej a una lista, tupla, etc

Es equivalente a un list comprehension que incluya un if.

### `filter()` o comprensión de lista

Las dos formas permiten filtrar elementos de una lista.

Con `filter()` y una función `lambda`:

```python
list(filter(lambda color: len(color) == 4, colores))
```

Con una comprensión de lista:

```python
[color for color in colores if len(color) == 4]
```

Para una condición sencilla, la comprensión de lista suele ser más clara y más directa.

Regla práctica:

- si la condición se lee fácilmente dentro de un `if`, usa comprensión de lista;
- si ya tienes una función preparada, `filter()` puede ser cómodo.

Ejemplo donde `filter()` puede tener sentido:

```python
list(filter(str.isalpha, palabras))
```

Con `lambda`, normalmente es preferible una comprensión de lista.

Pero cuando el objetivo es solo sumar, no hace falta construir una lista completa antes. Podemos usar un generador:

```python
sum(importe for importe in importes if importe > 100)
```

La idea es la misma, pero el generador produce los valores uno por uno. Es una forma habitual y elegante cuando trabajamos con `sum()`, `any()`, `all()`, `max()` o `min()`.

### Crea una función para evaluar si la longitud del nombre de un producto supera la media de longitudes de una lista y devuelve un mensaje personalizado para cada producto

````python
# Pregunta: ¿qué necesitará la función del ejercicio?
# Respuesta: calcular la longitud media y comparar cada producto con esa media.
# Separar el cálculo de la media en una función ayuda a reutilizar esa parte.
productos = ['Bicicleta', 'Patinete', 'Scooter', 'Skate']

def calcular_longitud_media(prod):
    longitudes = [len(p) for p in prod]
    media = sum(longitudes) / len(longitudes)
    return media

def productos_mas_largos_que_media(prod, media):
    for p in prod:
        if len(p) > media:
            print(f"Producto '{p}' tiene longitud mayor que la media {media}.")
        else:
            print(f"Producto '{p}' tiene longitud menor o igual que la media {media}.")

media_longitud = calcular_longitud_media(productos)
productos_mas_largos_que_media(productos, media_longitud)
````

````python
# Aquí lo hacemos todo en una sola función.
# Ventaja: el usuario solo llama a una función y no se preocupa de calcular la media antes.
# Limitación: la función imprime directamente; si queremos reutilizar los mensajes, conviene devolverlos con return.
productos = ['Bicicleta', 'Patinete', 'Scooter', 'Skate']

def evaluar_longitud_productos(prod):
    longitudes = [len(p) for p in prod]
    media = sum(longitudes) / len(longitudes)
    print(f"Longitud media de los productos: {media}")

    for p in prod:
        if len(p) > media:
            print(f"Producto '{p}' tiene longitud mayor que la media.")
        else:
            print(f"Producto '{p}' tiene longitud menor o igual que la media.")

evaluar_longitud_productos(productos)
````

````python
# Codex solution à analiser
# Alternativa: devolver una lista de mensajes en vez de imprimir dentro de la función.
productos = ['Bicicleta', 'Patinete', 'Scooter', 'Skate']

def evaluar_longitud_productos(productos):
    media = sum(len(producto) for producto in productos) / len(productos)
    mensajes = []

    for producto in productos:
        if len(producto) > media:
            mensajes.append(f"Producto '{producto}' tiene longitud mayor que la media.")
        else:
            mensajes.append(f"Producto '{producto}' tiene longitud menor o igual que la media.")

    return mensajes

for mensaje in evaluar_longitud_productos(productos):
    print(mensaje)
````

### Nota Codex: `print()` o `return`

Cuando una función imprime con `print()`, muestra el resultado pero no lo devuelve para reutilizarlo.

Cuando una función usa `return`, el resultado se puede guardar, transformar, probar o imprimir después. Por eso, para funciones que calculan algo, suele ser mejor devolver datos y decidir fuera de la función cómo mostrarlos.
