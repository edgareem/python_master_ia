## 1. Leer errores sin miedo

Cuando Python falla, normalmente nos da tres pistas: el tipo de error, la linea aproximada y una descripcion. Aprender a leer errores es parte normal de programar.

```python
# Pregunta del estudiante: que hago cuando aparece NameError?
# Respuesta del profesor: NameError significa normalmente que Python no conoce ese nombre.
# En este caso, la variable precio no existe todavia.
# La linea esta comentada para que el notebook pueda ejecutarse completo sin detenerse.

# precio + 10
```

```python
# Pregunta del estudiante: como corrijo el NameError anterior?
# Respuesta del profesor: primero hay que crear la variable, y despues ya se puede usar.
# Python ejecuta de arriba hacia abajo, por eso el orden importa.

precio = 25
precio + 10
```

## 2. Operadores utiles: `%`, `//` y prioridad

`%` devuelve el resto de una division. `//` devuelve la division entera. Son muy utiles para condiciones y calculos discretos.

```python
# Pregunta del estudiante: para que sirve el operador %?
# Respuesta del profesor: % devuelve el resto de una division.
# Si numero % 2 da 0, el numero es divisible entre 2 y por tanto es par.

numero = 8
numero % 2 == 0
```

```python
# Pregunta del estudiante: que diferencia hay entre / y //?
# Respuesta del profesor: / hace division normal; // hace division entera.
# 17 / 5 seria 3.4, pero 17 // 5 devuelve 3.

17 // 5
```

```python
# Pregunta del estudiante: por que los dos resultados son distintos?
# Respuesta del profesor: Python hace primero multiplicaciones y divisiones.
# Los parentesis cambian el orden y tambien ayudan a leer la intencion.

resultado_1 = 2 + 3 * 4
resultado_2 = (2 + 3) * 4

print(resultado_1)
print(resultado_2)
```

## 3. Slices mas completos

Los slices tienen la forma `[inicio:final:paso]`. El final no se incluye. Esto sera muy importante en NumPy.

```python
# Pregunta del estudiante: como leo tantos slices sin confundirme?
# Respuesta del profesor: piensa siempre en posiciones y en la regla de que el final no entra.
# datos[0] es el primer elemento; datos[-1] empieza desde el final.
# datos[:3] omite el inicio, entonces empieza desde 0.
# datos[2:] omite el final, entonces llega hasta el final real.
# datos[::-1] usa paso negativo y recorre la lista al reves.

datos = [10, 20, 30, 40, 50, 60]

print(datos[0])
print(datos[-1])
print(datos[:3])
print(datos[2:])
print(datos[::2])
print(datos[::-1])
```

## 4. Booleanos: Python base vs NumPy/Pandas

En Python base se usan `and`, `or`, `not`. En NumPy/Pandas se usan mucho `&`, `|`, `~` para combinar condiciones elemento a elemento. Por ahora, lo importante es no mezclarlos sin pensar.

```python
# Pregunta del estudiante: cuando uso and?
# Respuesta del profesor: usa and cuando necesitas que todas las condiciones sean verdaderas.
# Aqui la persona debe tener al menos 18 anos Y tener permiso.

edad = 20
tiene_permiso = True

edad >= 18 and tiene_permiso
```

```python
# Pregunta del estudiante: si existe and, por que luego NumPy/Pandas usan &?
# Respuesta del profesor: and compara condiciones booleanas simples.
# En NumPy/Pandas se comparan muchos valores a la vez, por eso se usa & elemento a elemento.
# Los parentesis son importantes para separar claramente cada condicion.

x = 15
(x > 10) and (x < 20)
```

## 5. Valores que Python considera falsos

En un `if`, Python no solo entiende `True` y `False`. Tambien considera falsos valores como `0`, texto vacio, listas vacias y `None`.

```python
# Pregunta del estudiante: por que algunos valores entran en else si no son False literalmente?
# Respuesta del profesor: Python considera ciertos valores como falsos en condiciones.
# Son falsos: 0, texto vacio, lista vacia y None.
# Son verdaderos: numeros distintos de 0, textos no vacios y listas con elementos.

valores = [0, 1, '', 'hola', [], [1, 2], None]

for valor in valores:
    if valor:
        print(valor, '-> se comporta como True')
    else:
        print(valor, '-> se comporta como False')
```

## 6. `==` vs `is`

`==` compara valores. `is` compara identidad, es decir, si dos nombres apuntan exactamente al mismo objeto. Para comprobar `None`, se recomienda `is None`.

```python
# Pregunta del estudiante: por que se recomienda is None y no == None?
# Respuesta del profesor: None representa un objeto especial: ausencia de valor.
# Con None se comprueba identidad con is, porque queremos saber si es exactamente None.

respuesta = None

if respuesta is None:
    print('Todavia no hay respuesta')
```

## 7. Mutabilidad y copias

Las listas son mutables: se pueden modificar. Un error frecuente es pensar que `b = a` crea una copia, pero en realidad crea otro nombre para la misma lista.

```python
# Pregunta del estudiante: por que cambia a si yo solo modifique b?
# Respuesta del profesor: b = a no crea una copia; crea otro nombre para la misma lista.
# Como a y b apuntan al mismo objeto, modificar b tambien se ve en a.

a = [1, 2, 3]
b = a

b.append(4)

print(a)
print(b)
```

```python
# Pregunta del estudiante: como evito que una lista cambie cuando modifico otra?
# Respuesta del profesor: crea una copia independiente con .copy().
# Ahora b empieza con los mismos valores que a, pero vive como otra lista.

a = [1, 2, 3]
b = a.copy()

b.append(4)

print(a)
print(b)
```

## 8. Metodos inplace vs funciones que devuelven algo

Algunos metodos modifican el objeto original y no devuelven una nueva version. Esto se llama trabajar `inplace`.

```python
# Pregunta del estudiante: por que resultado imprime None?
# Respuesta del profesor: sort() modifica la lista original y no devuelve una lista nueva.
# A esto se le llama trabajar inplace.

numeros = [3, 1, 2]

resultado = numeros.sort()

print(numeros)
print(resultado)
```

```python
# Pregunta del estudiante: cuando uso sorted() en vez de sort()?
# Respuesta del profesor: usa sorted() si quieres una lista ordenada nueva sin cambiar la original.
# Usa sort() si quieres modificar la lista original directamente.

numeros = [3, 1, 2]
ordenados = sorted(numeros)

print(numeros)
print(ordenados)
```

## 9. `print()` vs `return`

`print()` muestra algo en pantalla. `return` devuelve un resultado para poder reutilizarlo. Esta diferencia es clave para escribir funciones.

```python
# Pregunta del estudiante: por que resultado es None si la funcion mostro 10?
# Respuesta del profesor: print() solo muestra algo en pantalla.
# Como la funcion no tiene return, no devuelve ningun valor reutilizable.

def mostrar_doble(numero):
    print(numero * 2)

resultado = mostrar_doble(5)
print('resultado:', resultado)
```

```python
# Pregunta del estudiante: por que ahora si puedo usar resultado + 10?
# Respuesta del profesor: return devuelve un valor real al lugar donde se llamo la funcion.
# Por eso podemos guardarlo y reutilizarlo en otra operacion.

def calcular_doble(numero):
    return numero * 2

resultado = calcular_doble(5)
print('resultado:', resultado)
print('puedo reutilizarlo:', resultado + 10)
```

## 10. Alcance de variables: local vs global

Una variable creada dentro de una funcion vive dentro de esa funcion. Esto ayuda a evitar mezclas accidentales.

```python
# Pregunta del estudiante: por que se imprimen dos mensajes distintos?
# Respuesta del profesor: la variable mensaje dentro de la funcion es local.
# No reemplaza la variable mensaje que esta fuera de la funcion.

mensaje = 'fuera'

def ejemplo_alcance():
    mensaje = 'dentro'
    return mensaje

print(ejemplo_alcance())
print(mensaje)
```

## 11. Argumentos posicionales y nombrados

Los argumentos posicionales dependen del orden. Los argumentos nombrados indican explicitamente que parametro recibe cada valor.

```python
# Pregunta del estudiante: que diferencia hay entre argumentos posicionales y nombrados?
# Respuesta del profesor: los posicionales dependen del orden; los nombrados dicen explicitamente que parametro recibe cada valor.
# Los nombrados suelen ser mas claros cuando hay varios parametros.

def calcular_precio(precio, descuento):
    return precio * (1 - descuento)

print(calcular_precio(100, 0.20))
print(calcular_precio(precio=100, descuento=0.20))
```

## 12. Cuidado con valores por defecto mutables

Evita usar listas o diccionarios como valores por defecto en funciones. Es un detalle avanzado, pero produce errores muy raros si no se conoce.

```python
# Pregunta del estudiante: por que lista=None y no lista=[]?
# Respuesta del profesor: una lista por defecto se reutilizaria entre llamadas y podria acumular datos sin querer.
# Con None creamos una lista nueva dentro de la funcion cada vez que hace falta.

def agregar_producto(producto, lista=None):
    if lista is None:
        lista = []
    lista.append(producto)
    return lista

print(agregar_producto('raton'))
print(agregar_producto('teclado'))
```

## 13. `enumerate()` para posicion y valor

Cuando necesitas la posicion y el valor al mismo tiempo, `enumerate()` evita crear contadores manuales.

```python
# Pregunta del estudiante: para que sirve enumerate()?
# Respuesta del profesor: sirve para obtener posicion y valor al mismo tiempo.
# Asi evitamos crear un contador manual.

colores = ['rojo', 'verde', 'azul']

for posicion, color in enumerate(colores):
    print(posicion, color)
```

## 14. `zip()` y listas de distinta longitud

`zip()` empareja por posicion, pero se detiene cuando termina la lista mas corta.

```python
# Pregunta del estudiante: por que Pedro no aparece en el resultado?
# Respuesta del profesor: zip() se detiene cuando termina la lista mas corta.
# Como productos solo tiene 2 elementos, solo se crean 2 parejas.

clientes = ['Ana', 'Luis', 'Pedro']
productos = ['Laptop', 'Tablet']

list(zip(clientes, productos))
```

## 15. List comprehensions para crear listas

Una list comprehension se usa normalmente para construir una lista nueva, no solo para ejecutar `print()`.

```python
# Pregunta del estudiante: como leo una list comprehension?
# Respuesta del profesor: de izquierda a derecha: que quiero producir, de donde saco los datos, y si hay condicion.
# dobles: produce numero * 2 por cada numero.
# pares: conserva numero solo si numero % 2 == 0.

numeros = [1, 2, 3, 4, 5]

dobles = [numero * 2 for numero in numeros]
pares = [numero for numero in numeros if numero % 2 == 0]

print(dobles)
print(pares)
```

## 18. Mini-practica final

Objetivo: usar varias piezas juntas antes de pasar a NumPy.

Tarea sugerida:

1. Crea una lista de importes.
2. Filtra los importes mayores o iguales a 100.
3. Calcula el total de esos importes.
4. Crea una funcion que devuelva un mensaje segun el total.
5. Usa `return`, no solo `print()`.

```python
# Pregunta del estudiante: que conceptos junta esta mini-practica?
# Respuesta del profesor: lista, filtro con comprehension, suma, funcion, if/else, return y print.
# Es un ejemplo pequeno de como combinar piezas basicas para resolver un problema.

importes = [50, 120, 80, 200, 30]

importes_altos = [importe for importe in importes if importe >= 100]
total = sum(importes_altos)

def evaluar_total(total):
    if total >= 300:
        return 'Total alto'
    else:
        return 'Total moderado'

print(importes_altos)
print(total)
print(evaluar_total(total))
```
