Soluciones propuestas para los ejercicios del complemento Codex. Hay varias formas validas de resolver muchos ejercicios; estas son versiones simples y didacticas.

## Ejercicio 1: Leer y corregir un error

```python
# Primero creamos la variable. Despues la usamos.
precio = 25
precio + 10
```

## Ejercicio 2: Par o impar con `%`

```python
numero = 17

# Un numero es impar si el resto al dividir entre 2 no es 0.
numero % 2 != 0
```

## Ejercicio 3: Division normal y division entera

```python
print(23 / 4)
print(23 // 4)
```

## Ejercicio 4: Prioridad de operadores

```python
# En la primera expresion Python calcula primero 5 * 3.
# En la segunda, los parentesis obligan a calcular primero 2 + 5.

print(2 + 5 * 3)
print((2 + 5) * 3)
```

## Ejercicio 5: Slices completos

```python
datos = [10, 20, 30, 40, 50, 60, 70]

print(datos[:3])
print(datos[3:])
print(datos[::2])
print(datos[::-1])
```

## Ejercicio 6: Condiciones combinadas

```python
edad = 16
tiene_autorizacion = True

puede_entrar = (edad >= 18) or tiene_autorizacion
puede_entrar
```

## Ejercicio 7: Valores falsy

```python
entradas = [0, 1, '', 'python', [], [5], None]

for entrada in entradas:
    if entrada:
        print(entrada, '-> True')
    else:
        print(entrada, '-> False')
```

## Ejercicio 8: `is None`

```python
respuesta = None

if respuesta is None:
    print('Sin respuesta todavia')
```

## Ejercicio 9: Mutabilidad y copias

```python
original = ['Ana', 'Luis']
copia = original.copy()

copia.append('Marta')

print(original)
print(copia)
```

## Ejercicio 10: `sort()` vs `sorted()`

```python
numeros = [4, 1, 3, 2]
ordenados = sorted(numeros)

print(numeros)
print(ordenados)
```

## Ejercicio 11: `return` reutilizable

```python
def calcular_total(precio, unidades):
    return precio * unidades

total = calcular_total(precio=12, unidades=5)
print(total + 10)
```

## Ejercicio 12: Alcance local y global

```python
estado = 'fuera'

def ver_estado():
    estado = 'dentro'
    return estado

print(ver_estado())
print(estado)
```

## Ejercicio 13: Argumentos nombrados

```python
def aplicar_descuento(precio, descuento):
    return precio * (1 - descuento)

aplicar_descuento(precio=100, descuento=0.15)
```

## Ejercicio 14: Valor por defecto seguro

```python
def agregar_tarea(tarea, lista=None):
    if lista is None:
        lista = []
    lista.append(tarea)
    return lista

print(agregar_tarea('estudiar'))
print(agregar_tarea('practicar'))
```

## Ejercicio 15: `enumerate()`

```python
productos = ['raton', 'teclado', 'monitor']

for posicion, producto in enumerate(productos):
    print(posicion, producto)
```

## Ejercicio 16: `zip()` con distinta longitud

```python
clientes = ['Ana', 'Luis', 'Pedro']
productos = ['Laptop', 'Tablet']

# Pedro no aparece porque zip se detiene en la lista mas corta.
list(zip(clientes, productos))
```

## Ejercicio 17: List comprehension para crear datos

```python
numeros = [1, 2, 3, 4, 5, 6]

cuadrados = [numero ** 2 for numero in numeros]
pares = [numero for numero in numeros if numero % 2 == 0]

print(cuadrados)
print(pares)
```

## Ejercicio 18: Mini-practica final

```python
ventas = [80, 120, 50, 300, 40]

ventas_altas = [venta for venta in ventas if venta >= 100]
total = sum(ventas_altas)

def evaluar_ventas(total):
    if total >= 400:
        return 'Buen resultado'
    else:
        return 'Resultado a revisar'

print(ventas_altas)
print(total)
print(evaluar_ventas(total))
```
