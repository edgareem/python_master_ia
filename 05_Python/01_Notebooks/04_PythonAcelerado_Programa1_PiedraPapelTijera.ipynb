**PASO 1:**

Define una función, llamada jugador, que pregunte al jugador qué jugada quiere hacer usando los códigos de abajo, y devuelva el resultado como un entero.

1: Piedra

2: Papel

3: Tijera

````python
def jugador():
    # Pedimos al usuario un código numérico.
    # input() siempre devuelve texto, por eso convertimos la respuesta con int().
    respuesta = input('\n ¿Qué jugada eliges (1: Piedra, 2: Papel, 3: Tijera)?: ')
    return int(respuesta)
````

### Nota Codex: validar la entrada del jugador

La función funciona si el usuario escribe `1`, `2` o `3`. El punto delicado es que `int(respuesta)` fallará si el usuario escribe texto, deja el campo vacío o pone un número fuera de rango.

Para aprender, esta versión está bien. En un programa más robusto, convendría repetir la pregunta hasta recibir una opción válida.

````python
# Codex solution à analiser
# Alternativa: validar la entrada antes de devolverla.
def jugador_validado():
    opciones_validas = {'1', '2', '3'}

    while True:
        respuesta = input('\n¿Qué jugada eliges (1: Piedra, 2: Papel, 3: Tijera)?: ')

        if respuesta in opciones_validas:
            return int(respuesta)

        print('Opción no válida. Escribe 1, 2 o 3.')
````

**PASO 2:**

Define una función, llamada ordenador, que seleccione aleatoriamente entre piedra, papel o tijera y devuelva el resultado.

Pista: necesitarás el módulo random, y puedes buscar una función que te ayude en:

https://www.w3schools.com/python/module_random.asp

````python
import random

def ordenador():
    # Guardamos las opciones posibles en una lista.
    # random.choice(lista) elige un elemento aleatorio de esa lista.
    opciones = ['piedra', 'papel', 'tijera']
    opcion = random.choice(opciones)
    return opcion
````

### Nota Codex: usar la misma representación

Aquí el ordenador devuelve texto (`'piedra'`, `'papel'`, `'tijera'`), mientras que el jugador devuelve números (`1`, `2`, `3`). Eso obliga a comparar números contra textos en la función `compara()`.

Una mejora importante es transformar también la jugada del jugador en texto. Así ambos lados hablan el mismo idioma: texto contra texto.

**PASO 3:**

Define una función, llamada compara, que tome como parámetros el resultado de la jugada del jugador y del ordenador, implemente la lógica del juego y devuelva el resultado, que podrá ser quien ha ganado o si ha habido empate.

````python
def compara(jugador, ordenador):
    # Esta función recibe dos cosas:
    # - jugador: un número entero, 1, 2 o 3
    # - ordenador: un texto, 'piedra', 'papel' o 'tijera'
    # Funciona, pero mezcla dos formas distintas de representar la misma idea.
    ganador = None

    if jugador == 1 and ordenador == 'tijera':
        ganador = 'jugador'
    elif jugador == 1 and ordenador == 'papel':
        ganador = 'ordenador'
    elif jugador == 1 and ordenador == 'piedra':
        ganador = 'empate'
    elif jugador == 2 and ordenador == 'tijera':
        ganador = 'ordenador'
    elif jugador == 2 and ordenador == 'papel':
        ganador = 'empate'
    elif jugador == 2 and ordenador == 'piedra':
        ganador = 'jugador'
    elif jugador == 3 and ordenador == 'tijera':
        ganador = 'empate'
    elif jugador == 3 and ordenador == 'papel':
        ganador = 'jugador'
    elif jugador == 3 and ordenador == 'piedra':
        ganador = 'ordenador'
    else:
        print('Ha habido algún error')

    return ganador
````

### Nota Codex: simplificar la lógica de victoria

La función `compara()` es correcta, pero tiene muchos `if` y `elif`. Eso se vuelve difícil de mantener.

Una forma más clara es guardar las reglas en un diccionario o conjunto:

```python
reglas = {
    'piedra': 'tijera',
    'papel': 'piedra',
    'tijera': 'papel'
}
```

La lectura sería: **la clave gana al valor**. Por ejemplo, `piedra` gana a `tijera`.

````python
# Codex solution à analiser
# Alternativa: comparar usando texto contra texto y un diccionario de reglas.
def compara_simple(jugada_jugador, jugada_ordenador):
    # Diccionario con las reglas del juego.
    # La clave gana al valor:
    # - 'piedra' gana a 'tijera'
    # - 'papel' gana a 'piedra'
    # - 'tijera' gana a 'papel'
    reglas = {
        'piedra': 'tijera',
        'papel': 'piedra',
        'tijera': 'papel',
    }

    # Si los dos sacan lo mismo, no gana nadie.
    if jugada_jugador == jugada_ordenador:
        return 'empate'

    # Buscamos qué jugada puede vencer la jugada del jugador.
    # Por ejemplo, si jugada_jugador es 'piedra',
    # reglas['piedra'] devuelve 'tijera'.
    #
    # Si el ordenador ha sacado justo esa jugada ('tijera'), entonces
    # significa que el jugador gana.
    if reglas[jugada_jugador] == jugada_ordenador:
        return 'jugador'

    # Si no hay empate y el jugador no gana,
    # entonces necesariamente gana el ordenador.
    return 'ordenador'
````

**PASO 4:**

Define la parte principal del programa que vaya siguiendo lo definido en el diseño y llamando a las funciones que hemos creado cuando sea necesario.

Cuando lo hayas terminado ejecuta todas las celdas y diviértete!!

````python
nombre = input('Cómo te llamas?: ')
print(f'Hola {nombre}, vamos a echar 10 jugadas. ¡Y el que consiga más puntos gana!')

# Contadores de puntos.
puntos_j = 0
puntos_o = 0

# range(10) repite el bloque 10 veces: una vuelta por jugada.
for cada in range(10):
    tirada_jugador = jugador()
    tirada_ordenador = ordenador()
    ganador = compara(tirada_jugador, tirada_ordenador)

    if ganador == 'jugador':
        print(f'\nYo he sacado {tirada_ordenador}')
        print('\nAsí que esta la has ganado tú!!')
        puntos_j += 1
    elif ganador == 'ordenador':
        print(f'\nYo he sacado {tirada_ordenador}')
        print('\nAsí que esta la he ganado yo!!')
        puntos_o += 1
    elif ganador == 'empate':
        print(f'\nVaya, yo también he sacado {tirada_ordenador}')
        print('\nAsí que hemos empatado, venga otra mano')
    else:
        print('Ha habido algún problema')

if puntos_j > puntos_o:
    print('\nHEMOS TERMINADO!!')
    print(f'\nEl resultado final es que has ganado tú con {puntos_j} puntos')
elif puntos_j < puntos_o:
    print('\nHEMOS TERMINADO!!')
    print(f'\nEl resultado final es que he ganado yo con {puntos_o} puntos')
else:
    print('\nHEMOS TERMINADO!!')
    print('\nPues parece que hemos empatado, así que todos ganamos!')
````

### Nota Codex: separar la lógica del juego

La parte principal funciona, pero mezcla varias responsabilidades:

Para un primer programa está bien. Para mejorar, conviene crear una función `jugar_partida()` que controle la partida completa. Eso hace el programa más fácil de leer, probar y modificar.

## Version Codex

Esta versión resume las mejoras importantes para estudiar:

- validar la entrada del jugador;
- usar la misma representación para jugador y ordenador;
- guardar las reglas del juego en un diccionario;
- separar la partida en funciones más pequeñas.

````python
# Codex solution à analiser
import random

OPCIONES = {
    '1': 'piedra',
    '2': 'papel',
    '3': 'tijera',
}

REGLAS = {
    'piedra': 'tijera',
    'papel': 'piedra',
    'tijera': 'papel',
}


def pedir_jugada():
    while True:
        respuesta = input('\n¿Qué jugada eliges (1: Piedra, 2: Papel, 3: Tijera)?: ')

        if respuesta in OPCIONES:
            return OPCIONES[respuesta]

        print('Opción no válida. Escribe 1, 2 o 3.')


def jugada_ordenador():
    return random.choice(list(OPCIONES.values()))


def comparar(jugada_jugador, jugada_pc):
    if jugada_jugador == jugada_pc:
        return 'empate'

    if REGLAS[jugada_jugador] == jugada_pc:
        return 'jugador'

    return 'ordenador'


def jugar_partida(rondas=10):
    nombre = input('Cómo te llamas?: ')
    print(f'Hola {nombre}, vamos a echar {rondas} jugadas. ¡Quien consiga más puntos gana!')

    puntos_jugador = 0
    puntos_ordenador = 0

    for ronda in range(1, rondas + 1):
        print(f'\nRonda {ronda}')
        jugada_jugador = pedir_jugada()
        jugada_pc = jugada_ordenador()
        ganador = comparar(jugada_jugador, jugada_pc)

        print(f'Tú has sacado {jugada_jugador}.')
        print(f'Yo he sacado {jugada_pc}.')

        if ganador == 'jugador':
            print('Esta la has ganado tú.')
            puntos_jugador += 1
        elif ganador == 'ordenador':
            print('Esta la he ganado yo.')
            puntos_ordenador += 1
        else:
            print('Empate.')

    print('\nHEMOS TERMINADO!!')
    print(f'Resultado final: {puntos_jugador} - {puntos_ordenador}')

    if puntos_jugador > puntos_ordenador:
        print(f'Has ganado tú, {nombre}!')
    elif puntos_jugador < puntos_ordenador:
        print('He ganado yo.')
    else:
        print('Hemos empatado.')


# Ejecuta esta línea para jugar:
# jugar_partida()
````
