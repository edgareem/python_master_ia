**PASO 1:**

Crea un diccionario vacío llamado agenda y define, siguiendo el diseño indicado, funciones para:

* Crear un registro

* Borrar un registro

* Modificar un registro

````python
# La agenda será un diccionario.
# Clave: nombre del contacto.
# Valor: teléfono del contacto.
agenda = {}
````

### Nota Codex: por qué un diccionario

Un diccionario es una buena estructura para una agenda porque permite asociar una clave con un valor:

```python
agenda['Ana'] = '612345678'
```

Se lee así: el contacto `Ana` tiene el teléfono `612345678`.

La ventaja es que luego podemos buscar, modificar o borrar usando directamente el nombre.

````python
def crear_registro():
    # Pedimos los datos del nuevo contacto.
    nombre = input('Nombre: ')
    tlf = input('Teléfono: ')

    # En un diccionario, agenda[nombre] = tlf crea una clave nueva
    # o reemplaza el valor si ese nombre ya existía.
    agenda[nombre] = tlf
````

### Nota Codex: variable global `agenda`

La función usa `agenda` aunque `agenda` no está creada dentro de la función. Eso funciona porque `agenda` existe fuera y es un diccionario mutable.

Para aprender está bien, pero en programas más claros suele ser mejor pasar la agenda como parámetro:

```python
def crear_registro(agenda):
    ...
```

Así la función muestra claramente qué datos necesita para trabajar.

````python
# Codex solution à analiser
# Alternativa: pasar la agenda como parámetro y devolverla modificada.
def crear_registro_mejorado(agenda):
    nombre = input('Nombre: ').strip() # strip() elimina espacios al principio y al final
    telefono = input('Teléfono: ').strip()

    agenda[nombre] = telefono
    print('Registro creado con éxito!')

    return agenda
````

````python
def borrar_registro():
    # Pedimos la clave que queremos borrar del diccionario.
    a_borrar = input('Nombre a borrar: ')

    # del elimina la clave y su valor.
    # Si el nombre no existe, Python lanzará un KeyError.
    del agenda[a_borrar]
    print('Borrado con éxito!')
````

### Nota Codex: borrar una clave que no existe

Esta versión funciona si el nombre existe en la agenda. Pero si intentamos borrar un contacto que no existe, esta línea falla:

```python
del agenda[a_borrar]
```

Una forma más robusta es comprobar antes:

```python
if a_borrar in agenda:
    del agenda[a_borrar]
```

Esto evita un `KeyError`.

````python
# Codex solution à analiser
# Alternativa: comprobar que el contacto existe antes de borrarlo.
def borrar_registro_mejorado(agenda):
    nombre = input('Nombre a borrar: ').strip()

    if nombre in agenda:
        del agenda[nombre]
        print('Borrado con éxito!')
    else:
        print('Ese contacto no existe en la agenda.')

    return agenda
````

````python
def modificar_registro():
    # Pedimos el nombre del contacto que queremos modificar.
    nombre_modificar = input('Nombre a modificar: ')
    tlf_modificar = input('Nuevo teléfono: ')

    # Esta línea modifica el teléfono si el nombre existe.
    # Si no existe, crea un contacto nuevo con ese nombre.
    agenda[nombre_modificar] = tlf_modificar
````

### Nota Codex: modificar no es lo mismo que crear

Con esta línea:

```python
agenda[nombre_modificar] = tlf_modificar
```

Python modifica el teléfono si el contacto existe, pero también crea un contacto nuevo si no existe.

Eso puede ser útil en algunos casos, pero aquí el menú dice **modificar un registro existente**. Por eso conviene comprobar primero si el nombre está en la agenda.

````python
# Codex solution à analiser
# Alternativa: solo modificar si el contacto ya existe.
def modificar_registro_mejorado(agenda):
    nombre = input('Nombre a modificar: ').strip()

    if nombre in agenda:
        telefono = input('Nuevo teléfono: ').strip()
        agenda[nombre] = telefono
        print('Registro modificado con éxito!')
    else:
        print('Ese contacto no existe. Primero debes crearlo.')

    return agenda
````

En una versión más robusta, podemos leer la opción como texto y comprobar si está entre las opciones válidas: `'1'`, `'2'`, `'3'`, `'4'`.

## Version Codex

Esta versión resume las mejoras más importantes para estudiar:

- pasar `agenda` como parámetro a las funciones;
- validar que un contacto existe antes de borrar o modificar;
- validar la opción del menú;
- usar `.strip()` para limpiar espacios escritos por error;
- separar el menú principal en una función.

````python
# Codex solution à analiser

def crear_registro(agenda):
    nombre = input('Nombre: ').strip() 
    telefono = input('Teléfono: ').strip()

    agenda[nombre] = telefono
    print('Registro creado con éxito!')

    return agenda


def borrar_registro(agenda):
    nombre = input('Nombre a borrar: ').strip()

    if nombre in agenda:
        del agenda[nombre]
        print('Borrado con éxito!')
    else:
        print('Ese contacto no existe en la agenda.')

    return agenda


def modificar_registro(agenda):
    nombre = input('Nombre a modificar: ').strip()

    if nombre in agenda:
        telefono = input('Nuevo teléfono: ').strip()
        agenda[nombre] = telefono
        print('Registro modificado con éxito!')
    else:
        print('Ese contacto no existe. Primero debes crearlo.')

    return agenda


def mostrar_menu():
    print('1. Crear un nuevo registro')
    print('2. Modificar un registro existente')
    print('3. Eliminar un registro')
    print('4. Salir')


def ejecutar_agenda():
    agenda = {}
    opcion = ''

    while opcion != '4':
        mostrar_menu()
        opcion = input('Qué quieres hacer: ').strip()

        if opcion == '1':
            crear_registro(agenda)
        elif opcion == '2':
            modificar_registro(agenda)
        elif opcion == '3':
            borrar_registro(agenda)
        elif opcion == '4':
            print('Saliendo de la agenda.')
        else:
            print('Opción no válida. Escribe 1, 2, 3 o 4.')

        print('\n')

    return agenda


# Ejecuta esta línea para usar la agenda:
# agenda_final = ejecutar_agenda()
````
