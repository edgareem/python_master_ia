### Nota Codex: `dtype`, bits y memoria

`dtype` significa **data type**, es decir, el tipo de dato que NumPy guarda dentro del array.

Por ejemplo, si ves:

```python
dtype('int64')
```

significa que cada entero usa 64 bits, es decir, 8 bytes.

Esto importa mucho en data science porque un array puede tener millones de elementos. Cambiar de `int64` a `int32` puede reducir la memoria a la mitad, siempre que los valores quepan en `int32`.

Que no nos confunda que Numpy sea una librería numérica. También puede haber arrays no numéricos.

Pero eso sí, **todos los componentes del array tienen que ser del mismo tipo** (que será una de las cosas que nos llevará a Pandas).

````python
# Codex solution à analiser
# Comparación de memoria entre int32 e int64.
a = np.array([1, 2, 3], dtype=np.int32)
b = np.array([1, 2, 3], dtype=np.int64)

print('a:', a.dtype, a.nbytes, 'bytes')
print('b:', b.dtype, b.nbytes, 'bytes')
````

````python
v = np.array(['a','b','c'])
type(v)
````

### Verificación Codex: enteros de tamaño fijo

A diferencia del `int` normal de Python, los enteros de NumPy tienen tamaño fijo. Eso permite eficiencia, pero también puede producir **overflow** si el resultado no cabe en el tipo.

Para ver el rango de un tipo entero:

```python
np.iinfo(np.int32)
np.iinfo(np.int64)
```

````python
#Si ponemos de diferentes tipos vemos que los convierte para que sean del mismo tipo
v = np.array(['a',2,'c'])
v
````

### Nota Codex: NumPy busca un tipo común

Un array de NumPy intenta guardar todos sus elementos con el mismo `dtype`.

Por eso, si mezclas texto y números:

```python
np.array(['a', 2, 'c'])
```

NumPy convierte el `2` en texto. Esto es muy importante: un array mezclado puede dejar de ser numérico y perder la capacidad de hacer operaciones matemáticas normales.

### Nota Codex: `arange()` vs `linspace()`

Las dos funciones crean secuencias, pero responden a preguntas distintas.

`np.arange(inicio, fin, paso)`:

- controlas el salto entre valores;
- el final normalmente no se incluye.

`np.linspace(inicio, fin, cantidad)`:

- controlas cuántos valores quieres;
- por defecto incluye el inicio y el final.

Regla práctica:

- si te importa el **paso**, usa `arange()`;
- si te importa la **cantidad exacta de puntos**, usa `linspace()`.

### Nota Codex: semilla y reproducibilidad

La semilla permite repetir los mismos números aleatorios. Esto es muy útil para aprender, depurar y comparar resultados.

En código moderno de NumPy, se recomienda usar un generador explícito:

```python
rng = np.random.default_rng(1234)
```

Las funciones antiguas como `np.random.rand()` siguen funcionando, pero `default_rng()` es mejor hábito para proyectos.

````python
# Codex solution à analiser
# Alternativa moderna con default_rng().
rng = np.random.default_rng(1234)

print("Array de 5x10 con números aleatorios entre 0 y 1:\n")
print(rng.random((5, 10))) # genera un array de 5x10 con números aleatorios entre 0 y 1
print("\n Array de 5 números aleatorios con distribución normal:")
print("\n" + str(rng.normal(size=5))) # genera un array de 5 números aleatorios con distribución normal
print("\n Array de 10 números aleatorios entre 1 y 10:")
print("\n" + str(rng.integers(1, 11, size=10))) # genera un array de 10 números aleatorios entre 1 y 10
````

### Nota Codex: condición para hacer `reshape()`

Para cambiar la forma de un array, el número total de elementos tiene que cuadrar.

Ejemplo:

```python
v = np.arange(20)
```

Tiene 20 elementos. Entonces puede pasar a:

```python
(2, 10)  # 20 elementos
(4, 5)   # 20 elementos
(5, 4)   # 20 elementos
```

Pero no puede pasar a `(3, 7)`, porque eso necesitaría 21 elementos.

````python
# Codex solution à analiser
# Usar -1 permite que NumPy calcule automáticamente una dimensión.
v = np.arange(20)

print(v.reshape(4, -1))  # 4 filas, columnas calculadas automáticamente
print("\n")
print(v.reshape(-1, 5))  # filas calculadas automáticamente, 5 columnas
````

### Nota Codex: `sort()` modifica el array original

`v.sort()` ordena el array **in-place**, es decir, modifica directamente `v`.

Si quieres obtener una versión ordenada sin tocar el array original, usa:

```python
np.sort(v)
```

Esta diferencia es muy importante porque muchas funciones de Python/NumPy pueden modificar el objeto original.

````python
# Codex solution à analiser
# Comparación entre v.sort() y np.sort(v).
v = np.array([3, 1, 2])
ordenado = np.sort(v)

print('original:', v)
print('ordenado:', ordenado)
````

### Nota Codex: estadísticos con `axis`

En el ejemplo se calculan estadísticos sobre un vector. En matrices, normalmente queremos calcular por filas o por columnas.

Recuerda:

- `axis=0`: resume por columnas;
- `axis=1`: resume por filas.

Esto aparece constantemente en data science.

````python
# Codex solution à analiser
# Estadísticos por filas y columnas.
m = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print('media total:', m.mean())
print('media por columnas:', m.mean(axis=0))
print('media por filas:', m.mean(axis=1))
````

### Nota Codex: `argmax()` devuelve posición, no valor

`v.max()` devuelve el valor máximo.

`v.argmax()` devuelve la posición del valor máximo.

Por eso para recuperar el valor usando la posición escribimos:


### CREAR COPIAS

````python
v1 = np.arange(5)
print(v1)
````

````python
v2 = v1
v2
````

````python
v2[:] = 5
````

````python
v2
````

````python
# v1 se modifica también porque v2 es una vista de v1, no una copia. Para crear una copia, se puede usar el método copy().
v1
````

````python
#Para que no modifique el original hay que decir explícitamente que quieres una copia
#Si no lo entenderá como un puntero al original (parar ahorrar memoria) y lo podrá modificar
v1 = np.arange(5)
print("v1 =", v1)
print("\n")
v2 = v1.copy() # aquí se hace una copia de v1 y se asigna a v2, por lo que v2 es independiente de v1
v2[:] = 5 # ahora cambio v2 a 5, pero v1 no se verá afectado
print("v1 =", v1)
print("\n")
print("v2 =", v2)
````

### Verificación Codex: copia y vista

Cuando haces:

```python
v2 = v1
```

no creas un array nuevo. Creas otro nombre que apunta al mismo array.

Cuando haces:

```python
v2 = v1.copy()
```

sí creas un array independiente.

Esta diferencia evita muchos errores cuando se preparan datos para análisis o machine learning.

````python
# Codex solution à analiser
# Slicing básico suele crear una vista, no una copia.
v1 = np.arange(6)
print('v1:', v1)
v2 = v1[1:4] # es una vista de v1 sin el primer y último elemento, !no una copia!

v2[0] = 99

print('v1:', v1)
print('v2:', v2)
````

### Nota Codex: combinar filtros booleanos

Para una sola condición, esto funciona:

```python
v[v < 0.5]
```

Para varias condiciones, usamos `&`, `|` y paréntesis:

```python
v[(v > 0.2) & (v < 0.8)]
```

No usamos `and` porque NumPy compara elemento por elemento.

### Crea dos arrays aleatorios del mismo tamaño y calcula su matriz de correlación

#### Rappel : coefficient et matrice de corrélation

Le coefficient de corrélation mesure la relation linéaire entre deux variables :

- proche de `1` : les deux variables augmentent ensemble ;
- proche de `-1` : quand une variable augmente, l'autre diminue ;
- proche de `0` : il n'y a pas de relation linéaire claire.

`np.corrcoef(a, b)` renvoie une matrice de corrélation :

```python
[[corr(a, a), corr(a, b)],
 [corr(b, a), corr(b, b)]]
```

On a toujours `1` sur la diagonale car une variable est parfaitement corrélée avec elle-même : `corr(a, a) = 1` et `corr(b, b) = 1`.

````python

# On crée deux tableaux de 10 valeurs aléatoires.
a = np.random.rand(10)
b = np.random.rand(10)

print("a:", a)
print("\nb:", b)

# [0, 1] donne la corrélation entre a et b.
print("\ncoeficiente de correlación:", np.corrcoef(a,b)[0,1])

# La matrice complète contient aussi corr(a,a) et corr(b,b), qui valent toujours 1.
print("\ncoeficiente de correlación (matriz completa):\n", np.corrcoef(a,b))
````
