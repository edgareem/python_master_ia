Como decíamos al principio muchos de los algoritmos de machine learning se construyen con operaciones avanzadas sobre estructuras de datos como vectores o matrices.

**Un poco de teoría: profundizando en la comprensión de las estructuras de datos**

Hasta ahora hemos trabajado con tres tipos fundamentales de estructuras:

**Escalares** : en la lección de Python los llamábamos "datos individuales". Son valores simples, como un número, y se pueden asociar geométricamente a un punto. Por eso, decimos que tienen dimensión 0.

**Vectores** : son secuencias de escalares ordenadas. Geométricamente, un vector se representa como una flecha en el espacio y vive en un espacio de dimensión n, según el número de componentes. Por ejemplo, un vector de tres elementos vive en R3. Estructuralmente, también se pueden ver como matrices unidimensionales, ya sea de forma vertical (m×1) o horizontal (1×m).

**Matrices**: son colecciones de vectores organizados en filas y columnas. Desde un punto de vista geométrico, dos vectores independientes en Rn pueden definir un plano. Por eso, una matriz puede representar objetos bidimensionales (2D), y su forma se denota como m×n, con mm filas y nn columnas.

Podemos extender esta idea a estructuras más complejas, como los arrays multidimensionales (o tensores), que representan hiperplanos en espacios de dimensión superior. Estos aparecen, por ejemplo, en ciencia de datos, imágenes, o redes neuronales, y permiten trabajar con información en más de dos dimensiones.


 #### Nota Codex - Leer las formas antes de calcular
 
 En algebra lineal, casi siempre conviene mirar primero la **forma** de los objetos: escalar, vector, matriz o tensor. En NumPy esa forma se consulta con .shape. Si entiendes la forma, puedes anticipar si una operacion es posible antes de ejecutarla.
CodeCell

```python
# Ejemplo Codex: comprobar dimensiones con .shape
escalar = np.array(7)
vector = np.array([1, 2, 3])
matriz = np.array([[1, 2, 3], [4, 5, 6]])
tensor = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print('escalar shape:', escalar.shape) # Un escalar no tiene dimensiones, por lo que su shape es una tupla vacía. es un punto en el espacio.
print('vector shape:', vector.shape) # Un vector tiene una dimensión, por lo que su shape es una tupla con un solo elemento. es una línea en el espacio.
print('matriz shape:', matriz.shape) # Una matriz tiene dos dimensiones, por lo que su shape es una tupla con dos elementos. (usado en Álgebra Lineal para representar sistemas de ecuaciones, transformaciones lineales, tablas, etc.)
print('tensor shape:', tensor.shape) # Un tensor tiene tres dimensiones, por lo que su shape es una tupla con tres elementos. (usado en Deep Learning para representar imágenes, videos, etc.)
```

### OPERACIONES CON VECTORES

#### SUMAR VECTORES

La condición es que tienen que tener la misma longitud

#### Nota Codex - Sumar vectores no es juntar listas

Cuando sumamos vectores, no estamos concatenando elementos. Sumamos cada posicion con su posicion equivalente. Por eso [1, 2, 3] + [4, 5, 6] da [5, 7, 9], no [1, 2, 3, 4, 5, 6].


```python
# Creamos dos vectores con la misma longitud.
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

# NumPy suma componente a componente: [1+4, 2+5, 3+6].
v1 + v2
```

#### RESTAR VECTORES

La condición es que tienen que tener la misma longitud

```python
# La resta tambien se hace componente a componente: [1-4, 2-5, 3-6].
v1 - v2
```

#### MULTIPLICAR UN VECTOR POR UN ESCALAR

```python
# Multiplicar por un escalar cambia la magnitud del vector.
# Cada componente se multiplica por 10.
v1 * 10
```

#### MULTIPLICAR DOS VECTORES

La condición es que tienen que tener la misma longitud

El tipo de multiplicación más usado es el producto escalar o producto punto: multiplicar uno a uno los correspondientes y luego sumar los resultados.

#### Nota Codex - Intuicion del producto escalar
 
 El producto escalar resume en un solo numero cuanto se parecen dos vectores en direccion y magnitud. Si el resultado es grande y positivo, suelen apuntar en una direccion parecida. Si es cercano a cero, hay poca alineacion. Esta idea aparece mucho en recomendadores, busqueda semantica y modelos de machine learning.
CodeCell

```python
# Ejemplo Codex: producto escalar manual vs NumPy
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

producto_manual = (1 * 4) + (2 * 5) + (3 * 6)
producto_numpy = np.dot(v1, v2)

print(producto_manual)
print(producto_numpy)
```

```python
# Revisamos el primer vector antes de calcular el producto punto.
v1
```

```python
# Revisamos el segundo vector: debe tener la misma longitud que v1.
v2
```

```python
# Producto escalar con np.dot:
# 1*4 + 2*5 + 3*6 = 32.
print(np.dot(v1, v2))
```

```python
# El operador @ hace el mismo producto escalar entre dos vectores 1D.

# np.dot(v1, v2) y v1 @ v2 hacen lo mismo con dos vectores 1D.
# np.dot es mas explicito para aprender el concepto.
# @ es mas comodo cuando escribimos operaciones de algebra lineal.

print(v1 @ v2)
```

### OPERACIONES CON MATRICES

#### SUMAR MATRICES

La condición es que tienen que tener la misma forma

#### Nota Codex - Misma forma significa mismas coordenadas

Para sumar o restar matrices, cada posicion debe tener una posicion equivalente en la otra matriz. Por eso una matriz 2x3 se puede sumar con otra 2x3, pero no con una 3x2, aunque ambas tengan 6 numeros.


```python
# Dos matrices con la misma forma: 2 filas y 3 columnas.
m1 = np.array([[3, -4, 6], [0, -8, 3]])
m2 = np.array([[-1, -9, 7], [4, 8, -7]])

# La suma de matrices se hace posicion por posicion.
m1 + m2
```

```python
# Visualizamos m2: tiene la misma forma que m1.
m2
```

```python
# Repetimos la suma para reforzar la idea de operacion elemento a elemento.
m1 + m2
```

#### RESTAR MATRICES

La condición es que tienen que tener la misma forma

```python
# La resta de matrices tambien se hace posicion por posicion.
m1 - m2
```

#### MULTIPLICAR UNA MATRIZ POR UN ESCALAR

```python
# Multiplicar una matriz por un escalar multiplica cada elemento por ese valor.
m1 * 10
```

#### MULTIPLICAR DOS MATRICES

Usaremos la forma del producto punto.

La condición es que el número de columnas de la primera tiene que ser el mismo que el número de filas de la segunda.

Más formalmente es multiplicar una m x n por una n x k.

Por ejemplo multiplicar una 3x2 por una 2x4.

Es así porque lo que hace el producto punto es ir multiplicando cada fila de la primera por cada columna de la segunda.

El resultado será una nueva matriz de m x k, ej 3x4.

Donde el producto punto de fila1 de la primera por columna 1 de la segunda será el resultado de la fila1,columna1 de la tercera.

#### Nota Codex - Regla de oro para multiplicar matrices
 
 Para multiplicar A @ B, mira las formas: (m, n) @ (n, k) = (m, k). Las dimensiones del centro deben coincidir. Las dimensiones de fuera anuncian la forma del resultado.
CodeCell

```python
# Ejemplo Codex: anticipar la forma antes de multiplicar
A = np.array([[1, 2, 3], [4, 5, 6]])      # forma 2x3
B = np.array([[1, 0], [0, 1], [1, 1]])    # forma 3x2

print('A:', A.shape)
print('B:', B.shape)
print('A @ B:', (A @ B).shape)
A @ B
```

```python
# shape nos devuelve la forma de la matriz: (filas, columnas).
m1.shape
```

```python
# Volvemos a mirar m1 antes de multiplicarla por otra matriz.
m1
```

```python
# Creamos una matriz 3x3.
# Esto permite calcular m1 @ m3 porque m1 es 2x3 y m3 es 3x3.
m3 = np.array([
    [8, 6, 2],
    [6, 3, 4],
    [2, -4, 6]
])
m3.shape
```

```python
# Resultado de (2x3) @ (3x3): la forma final sera 2x3.
m1 @ m3
```

#### APLANAR UNA MATRIZ

Habrá ocasiones en machine learning que aunque tengamos que usar una matriz el formato en el que tenemos que pasarle esa matriz a la clase que la procesará es un array unidimensional.

Aplanar una matriz (flatten en inglés) significa eso, pasarla de arrays multidimensionales a un vector, simplemente concatenando todas sus filas en una sola.

Lo hacemos con el método flatten de Numpy.

#### Nota Codex - Flatten conserva los datos, cambia la forma

Aplanar no cambia los valores, solo cambia como estan organizados. Es util cuando un algoritmo espera una sola fila de caracteristicas en vez de una matriz. En imagenes, por ejemplo, una matriz de pixeles puede transformarse en un vector de entrada.


```python
# Recordemos como es la matriz m3 antes de aplanarla.
m3
```

```python
# flatten convierte la matriz en un vector 1D recorriendo las filas de izquierda a derecha.
m3.flatten()
```

### Suma v1 más v2 y resta v3

### Multiplica m1 por un escalar (3) y luego suma m2

```python
# Primero multiplicamos cada elemento de m1 por 3.
# Luego sumamos m2 posicion por posicion.
m1 = np.array([[1, 2, 3], [4, 5, 6]])
m2 = np.array([[6, 5, 4], [3, 2, 1]])

m1 * 3 + m2
```

### Calcula el producto escalar entre estos dos vectores y luego multiplicalo por un escalar (2)

```python
# Calculamos el producto escalar entre v1 y v2.
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

producto = np.dot(v1, v2) # o v1 @ v2
print(producto)

# Despues multiplicamos ese resultado por el escalar 2.
print(producto * 2)
```

### ¿Se pueden multiplicar estas matrices? Anticipa la forma que tendrá el resultado

```python
# m1 tiene forma 2x3 y m2 tiene forma 3x2.
# Se pueden multiplicar porque las dimensiones internas coinciden: 3 y 3.
# El resultado tendra forma 2x2.
m1 = np.array([[1, 2, 3], [4, 5, 6]])
m2 = np.array([[7, 8], [9, 10], [11, 12]])

m1 @ m2
```

### Multiplica estas matrices y aplana el resultado

```python
# m2 es la matriz identidad: al multiplicar por ella, m1 no cambia.
m1 = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
m2 = np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1]])

producto = m1 @ m2

# Aplanamos el resultado para convertirlo en un vector 1D.
producto.flatten()
```

```python
# Ejemplo Codex: multiplicar por la identidad no cambia la matriz
A = np.array([[2, -1], [0, 3]])
I = np.eye(2)

print(A @ I)
print(I @ A)
```

#### Nota Codex - La matriz identidad
 
 La matriz identidad funciona como el numero 1 en la multiplicacion: si las dimensiones son compatibles, A @ I = A e I @ A = A. Por eso aparece mucho en algebra lineal, optimizacion y modelos estadisticos.

### 1. Norma de un vector

#### Nota Codex

La norma mide el tamano o longitud de un vector. En machine learning se usa para medir distancias, magnitudes de errores y regularizacion. La norma euclidea de [3, 4] es 5, igual que en el teorema de Pitagoras.


```python
# Ejemplo Codex: norma de un vector
v = np.array([3, 4]) # 3^2 + 4^2 = 9 + 16 = 25, sqrt(25) = 5
np.linalg.norm(v)
```

### 2. Distancia entre vectores

#### Nota Codex

La distancia euclidea entre dos vectores se calcula restando los vectores y midiendo la norma del resultado. Sirve, por ejemplo, para comparar observaciones o encontrar vecinos parecidos.


```python
# Ejemplo Codex: distancia entre dos puntos
p1 = np.array([1, 2])
p2 = np.array([4, 6])

np.linalg.norm(p2 - p1)
```

### 3. Transpuesta de una matriz

#### Nota Codex

La transpuesta intercambia filas por columnas. Si una matriz tiene forma 2x3, su transpuesta tiene forma 3x2. Se usa mucho para preparar multiplicaciones matriciales.


```python
# Ejemplo Codex: transpuesta
A = np.array([[1, 2, 3], [4, 5, 6]])

print(A.shape)
print(A.T.shape)
A.T
```

### 4. Determinante

#### Nota Codex

El determinante es un numero asociado a una matriz cuadrada. Si el determinante es cero, la matriz no tiene inversa. Geometricamente, indica como una transformacion cambia areas o volumenes.


```python
# Ejemplo Codex: determinante de una matriz cuadrada
A = np.array([[1, 2], [3, 4]])
np.linalg.det(A)
```

### 5. Matriz inversa

#### Nota Codex

La inversa de una matriz cuadrada es parecida al inverso de un numero. Si A_inv es la inversa de A, entonces A @ A_inv produce la matriz identidad. No todas las matrices tienen inversa.


```python
# Ejemplo Codex: inversa de una matriz
A = np.array([[1, 2], [3, 4]])
A_inv = np.linalg.inv(A)

print(A_inv)
print(A @ A_inv)
```

### 6. Sistemas de ecuaciones lineales

#### Nota Codex

Muchos problemas se escriben como A @ x = b: conocemos A y , y queremos encontrar x. En NumPy se resuelve con 
p.linalg.solve, que suele ser mejor que calcular la inversa manualmente.


```python
# Ejemplo Codex: resolver A @ x = b
A = np.array([[2, 1], [1, 3]])
b = np.array([8, 13])

x = np.linalg.solve(A, b)
x
```

### 7. Valores y vectores propios

#### Nota Codex

Un vector propio mantiene su direccion cuando una matriz lo transforma; solo cambia su escala. Los valores propios indican cuanto se estira o encoge. Son fundamentales en PCA, reduccion de dimensionalidad y analisis de estabilidad.


```python
# Ejemplo Codex: valores y vectores propios
A = np.array([[2, 0], [0, 3]])
valores, vectores = np.linalg.eig(A)

print('valores propios:', valores)
print('vectores propios:')
print(vectores)
```

### 8. Rango de una matriz

#### Nota Codex

El rango mide cuanta informacion independiente contiene una matriz. Si una fila o columna se puede construir combinando otras, no aporta una nueva dimension de informacion.


```python
# Ejemplo Codex: rango de una matriz
A = np.array([[1, 2], [2, 4]])
np.linalg.matrix_rank(A)
```

### 9. Broadcasting

#### Nota Codex

Broadcasting es la capacidad de NumPy para combinar arrays de formas distintas cuando son compatibles. Es muy util, pero tambien puede generar errores si no revisamos bien las dimensiones.


```python
# Ejemplo Codex: broadcasting
M = np.array([[1, 2, 3], [4, 5, 6]])
v = np.array([10, 20, 30])

# v se suma a cada fila de M.
M + v
```

### 10. Descomposiciones matriciales

#### Nota Codex

Las descomposiciones separan una matriz en piezas mas simples. PCA, compresion de imagenes, recomendadores y muchos modelos numericos usan ideas como SVD, QR o eigendescomposicion. No hace falta dominarlas al principio, pero conviene saber que existen.


```python
# Ejemplo Codex: SVD, una descomposicion muy usada
A = np.array([[1, 2], [3, 4], [5, 6]])
U, S, Vt = np.linalg.svd(A)

print('U:', U.shape)
print('S:', S.shape)
print('Vt:', Vt.shape)
print(S)
```
