## Ejercicio 2: lista de Python y suma con comprensión de listas

### Ejercicio a analizar

Crea una lista de Python con los números del 1 al 10. Después suma 5 a cada uno de sus elementos.


```python
# Solución Codex
# range(1, 11) genera los números desde 1 hasta 10.
lista = list(range(1, 11))

# En Python puro, una lista no se puede sumar directamente con 5.
# Por eso usamos una comprensión de listas.
lista_mas_5 = [cada + 5 for cada in lista]
lista_mas_5
```

#### Nota Codex

En Python puro, `lista + 5` no funciona porque `+` entre listas significa concatenar listas, no sumar un número a cada elemento. Para hacer una operación elemento a elemento, usamos una comprensión de listas o NumPy.


```python
# Solución alternativa Codex: usando un bucle clásico
resultado = []
for cada in lista:
    resultado.append(cada + 5)

resultado
```

## Ejercicio 11: crear y ordenar un vector

### Ejercicio a analizar

Crea un vector llamado `vector`, como el anterior, pero ordénalo.


```python
# Solución Codex
vector = np.random.randint(1, 101, 100)

# sort() ordena el array en el mismo objeto y devuelve None.
vector.sort()
vector
```

#### Nota Codex

Atención: `vector.sort()` modifica `vector` directamente, pero no devuelve el vector ordenado. Por eso `print(vector.sort())` imprime `None`. Si quieres una copia ordenada, usa `np.sort(vector)`.


```python
# Solución alternativa Codex: crear una copia ordenada
vector_original = np.random.randint(1, 101, 100)
vector_ordenado = np.sort(vector_original)

print('Original:', vector_original)
print('Ordenado:', vector_ordenado)
```

## Ejercicio 12: estadísticos básicos sobre `vector`

### Ejercicio a analizar

Sobre `vector`, calcula la media, mediana, desviación típica, varianza, máximo y mínimo.


```python
# Solución Codex
print('Media:', vector.mean())
print('Mediana:', np.median(vector))
print('Desviación típica:', vector.std())
print('Varianza:', vector.var())
print('Máximo:', vector.max())
print('Mínimo:', vector.min())
```

#### Nota Codex

La media resume el centro promedio. La mediana es el valor central cuando los datos están ordenados. La desviación típica y la varianza miden dispersión: cuánto se alejan los datos del centro.


```python
# Solución alternativa Codex: usando funciones de NumPy
print('Media:', np.mean(vector))
print('Mediana:', np.median(vector))
print('Desviación típica:', np.std(vector))
print('Varianza:', np.var(vector))
print('Máximo:', np.max(vector))
print('Mínimo:', np.min(vector))
```

## Ejercicio 14: extraer los números pares

### Ejercicio a analizar

Extrae de `vector` los elementos que son pares. Pista: puedes apoyarte en el operador módulo `%`.


```python
# Solución Codex
# Un número es par si el resto de dividirlo entre 2 es 0.
pares = vector[vector % 2 == 0]
pares
```

#### Nota Codex

Aqué usamos una máscara booleana: `vector % 2 == 0` produce una serie de `True` y `False`. NumPy conserva solo las posiciones donde la condición es `True`.


```python
# Ejemplo Codex: ver la máscara booleana en pequeño
mini_vector = np.array([1, 2, 3, 4, 5, 6])
mascara = mini_vector % 2 == 0

print(mascara)
print(mini_vector[mascara])
```

## Ejercicio 15: sumar dos vectores aleatorios

### Ejercicio a analizar

Crea dos vectores de enteros llamados `v1` y `v2`, cada uno de 5 elementos aleatorios entre 1 y 10, y súmalos.


```python
# Solución Codex
v1 = np.random.randint(1, 11, 5)
v2 = np.random.randint(1, 11, 5)

suma = v1 + v2

print('v1:', v1)
print('v2:', v2)
print('suma:', suma)
```

## Ejercicio 16: producto escalar

### Ejercicio a analizar

Multiplícalos usando el producto escalar.


```python
# Solución Codex
# np.dot es muy explícito para aprender el concepto de producto escalar.
producto_escalar = np.dot(v1, v2)
producto_escalar
```

#### Nota Codex

Con dos vectores 1D, `np.dot(v1, v2)` y `v1 @ v2` dan el mismo resultado. Para aprender, `np.dot` deja claro que estamos haciendo un producto escalar; para álgebra lineal más compacta, `@` es muy cómodo.


```python
# Solución alternativa Codex
v1 @ v2
```

Esta sección reúne complementos útiles para evitar errores frecuentes cuando trabajas con NumPy y álgebra lineal.


## Complemento 1: fijar la semilla aleatoria

#### Nota Codex

Cuando usamos números aleatorios, cada ejecución puede dar resultados diferentes. Para practicar o corregir ejercicios, a veces conviene fijar una semilla con `np.random.seed()` y así obtener siempre los mismos valores.

## Complemento 2: diferencia entre `sort()` y `np.sort()`

#### Nota Codex

`array.sort()` ordena el array original y devuelve `None`. `np.sort(array)` devuelve una copia ordenada y conserva el array original sin modificar.


```python
# Ejemplo Codex
datos = np.array([4, 1, 3, 2])
copia_ordenada = np.sort(datos)

print('datos original:', datos)
print('copia ordenada:', copia_ordenada)

datos.sort()
print('datos después de sort():', datos)
```

## Complemento 3: revisar formas antes de operar

#### Nota Codex

Muchos errores de NumPy vienen de dimensiones incompatibles. Antes de sumar, multiplicar o hacer `reshape`, revisa `.shape`.


```python
# Ejemplo Codex
A = np.ones((2, 3))
B = np.ones((3, 2))

print('A:', A.shape)
print('B:', B.shape)
print('A @ B:', (A @ B).shape)
```

## Complemento 4: broadcasting básico

#### Nota Codex

Broadcasting permite combinar arrays de formas distintas cuando NumPy puede extender una dimensión automáticamente. Es potente, pero hay que usarlo con cuidado.


```python
# Ejemplo Codex
M = np.array([[1, 2, 3], [4, 5, 6]])
v = np.array([10, 20, 30])

# v se suma a cada fila de M.
M + v
```

## Complemento 5: producto matricial frente a producto elemento a elemento

#### Nota Codex

`A * B` multiplica elemento a elemento. `A @ B` realiza producto matricial. Son operaciones diferentes y pueden dar resultados muy distintos.


```python
# Ejemplo Codex
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])

print('Elemento a elemento:')
print(A * B)

print('Producto matricial:')
print(A @ B)
```
