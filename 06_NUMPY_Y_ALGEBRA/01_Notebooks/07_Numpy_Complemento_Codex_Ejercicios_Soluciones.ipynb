# NUMPY COMPLEMENTO CODEX: EJERCICIOS SOLUCIONES

Soluciones propuestas para los ejercicios del complemento NumPy Codex. Intenta entender cada línea, especialmente el uso de `axis`, máscaras booleanas, broadcasting y `np.where()`.

```python
import numpy as np
```

## Ejercicio 1: shape, ndim y size

```python
m = np.arange(1, 21).reshape(4, 5)

print(m)
print('shape:', m.shape)
print('ndim:', m.ndim)
print('size:', m.size)
```

## Ejercicio 2: medias por filas y columnas

```python
m = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print('Media total:', m.mean())
print('Media por columnas:', m.mean(axis=0))
print('Media por filas:', m.mean(axis=1))
```

## Ejercicio 3: broadcasting

```python
ventas = np.array([
    [100, 120, 130, 150],
    [80, 90, 110, 140],
    [200, 210, 190, 220]
])

media_columnas = ventas.mean(axis=0)
ventas_centradas = ventas - media_columnas

print('Media por columnas:', media_columnas)
print(ventas_centradas)
```

## Ejercicio 4: máscara booleana con dos condiciones

```python
v = np.arange(1, 101)
resultado = v[(v > 20) & (v < 40)]
print(resultado)
```

## Ejercicio 5: máscara booleana con OR

```python
v = np.arange(1, 101)
resultado = v[(v < 10) | (v > 90)]
print(resultado)
```

## Ejercicio 6: pares con máscara booleana

```python
v = np.arange(1, 31)
pares = v[v % 2 == 0]
print(pares)
```

## Ejercicio 7: `np.where()` con etiquetas

```python
notas = np.array([3, 5, 8, 4, 9, 6])
resultado = np.where(notas >= 5, 'aprobado', 'suspenso')
print(resultado)
```

## Ejercicio 8: `np.where()` con números

```python
ventas = np.array([80, 120, 300, 50, 200])
comision = np.where(ventas >= 100, ventas * 0.10, 0)
print(comision)
```

## Ejercicio 9: vectorización

```python
precios = np.array([100, 250, 80, 400])
precios_con_iva = precios * 1.21
print(precios_con_iva)
```

## Ejercicio 10: funciones universales

```python
v = np.array([1, 4, 9, 16, 25])
print(np.sqrt(v))
```

## Ejercicio 11: `np.clip()`

```python
v = np.array([-5, 0, 3, 10, 20])
print(np.clip(v, 0, 10))
```

## Ejercicio 12: reshape y transpuesta

```python
v = np.arange(1, 13)
m = v.reshape(3, 4)

print(m)
print(m.T)
```

## Ejercicio 13: `np.newaxis`

```python
v = np.array([1, 2, 3])

print('Original:', v.shape)
print('Fila:', v[np.newaxis, :].shape)
print('Columna:', v[:, np.newaxis].shape)
```

## Ejercicio 14: concatenar arrays

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate([a, b]))
print(np.vstack([a, b]))
print(np.hstack([a, b]))
```

## Ejercicio 15: dtype y astype

```python
v = np.array([1, 2, 3])
print(v.dtype)

v_float = v.astype(float)
print(v_float)
print(v_float.dtype)
```

## Ejercicio 16: valores `nan`

```python
v = np.array([10, 20, np.nan, 40])

print(np.isnan(v))
print(np.mean(v))
print(np.nanmean(v))
```

## Ejercicio 17: valores infinitos

```python
v = np.array([1, 2, np.inf, 4])
resultado = v[~np.isinf(v)]
print(resultado)
```

## Ejercicio 18: norma de un vector

```python
v = np.array([3, 4])
print(np.linalg.norm(v))
```

## Ejercicio 19: resolver sistema lineal

```python
A = np.array([
    [2, 1],
    [1, 3]
])
b = np.array([8, 13])

solucion = np.linalg.solve(A, b)
print(solucion)
```

## Ejercicio 20: reto final

```python
np.random.seed(123)

m = np.random.randint(1, 101, size=(5, 4))
media_columnas = m.mean(axis=0)
m_centrada = m - media_columnas
valores_altos = m[m > 50]
etiquetas = np.where(m > 50, 'alto', 'bajo')

print('Matriz original:')
print(m)
print('Media por columnas:')
print(media_columnas)
print('Matriz centrada:')
print(m_centrada)
print('Valores mayores que 50:')
print(valores_altos)
print('Etiquetas:')
print(etiquetas)
```
