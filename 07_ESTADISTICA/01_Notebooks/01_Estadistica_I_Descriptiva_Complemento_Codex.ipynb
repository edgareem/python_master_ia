## CONCEPTOS BÁSICOS

### Descriptiva Vs Inferencial

Son las 2 grandes ramas de la estadística.

Y por tanto así también vamos a organizar este módulo.

La **DESCRIPTIVA** se centra en describir  los datos que estamos analizando. Resumirlos, visualizarlos y sacar conclusiones.

Es la que más se usa en la práctica, y son el tipo de operaciones más básicas e imprescindibles que vamos a estar haciendo en Data Science.

Simplemente usando las técnicas que vamos a aprender en este apartado ya podremos hacer el 80% de los análisis que normalmente se hacen en contextos empresariales.

La **INFERENCIAL** se pregunta "vale, pero ¿esas conclusiones que hemos detectado en la descriptiva realmente son robustas y estables y pueden ser extrapoladas?".

Parte del concepto teórico de que realmente siempre vamos a estar analizando muestras, y no toda la población (más sobre esto en el siguiente apartado). 

Por lo que son necesarias técnicas que nos garanticen estadísticamente que esas conclusiones no son fruto del azar o de nuestra muestra en concreto.

Esta parte está muy infrautilizada en la práctica del día a día, ya que pocas veces se usan estas técnicas para contrastar si los resultados de un análisis, estudio o informe son realmente estadísticamente significativos.

Sin embargo como Data Scientist, conocerlas te va a servir tanto para garantizar que la calidad de tu trabajo es la correcta como para poder identificar falsas conclusiones que te intenten "colar".

Ahora bien, debes saber un hecho importante que afecta a todo lo que veamos en la parte inferencial.

Todas las técnicas y herramientas que veremos están basadas en que la muestra ha sido obtenida ALEATORIAMENTE y sin ningún tipo de sesgo.

Para garantizar esto la teoría nos dice que debemos hacer un correcto "diseño de experimento" que en la empresa lógicamente no se puede hacer casi nunca.

Así te corresponde como analista el evaluar el grado de "aleatoridad" y no sesgo que tienen tus datos y hasta qué punto puedes apoyarte en los recursos de la estadística inferencial.

### Escalas de medida y tipos de variables


#### Nota Codex - Precisión sobre tipos de variables

Antes de elegir una estadística o un gráfico, identifica el tipo de variable:

- **Cualitativa nominal**: categorías sin orden, por ejemplo sexo, país o color.
- **Cualitativa ordinal**: categorías con orden, por ejemplo bajo, medio, alto.
- **Cuantitativa discreta**: conteos, por ejemplo número de hijos o compras.
- **Cuantitativa continua**: medidas, por ejemplo altura, precio o tiempo.

Esta decisión manda sobre todo lo demás: qué resumen usar, qué gráfico hacer y qué test estadístico tiene sentido.


Este es uno de los puntos más importantes que debes conocer y sin embargo pocas veces es tratado en las formaciones de Data Science.

La estadística nos dice que hay 4 grandes escalas de medida:

1. **Nominal**: Tiene la propiedad de identidad (igual/diferente) pero NO la de orden. Ejemplos: sexo, colores, ...
2. **Ordinal**: Añade la propiedad de orden, pero NO tiene una igual distancia entre elementos. Ejemplos: clase social, nivel formativo, ...
3. **Intervalo (o discreta)**: Añade la propiedad de igualdad de distancias, pero no permite infinitos valores entre sus elementos. Ejemplos: número de hijos, número de siniestros, ...
4. **Razón (o continua)**: Puede tener (teóricamente) infinitos valores entre sus elementos. Ejemplos: precio de un producto, distancias, ...

A partir de las escalas se derivan los tipos de variables, que pueden ser diferente en cada implementación (por ej en cada lenguaje de programación), pero casi siempre podremos hacer una asignación aproximada a una escala.

Por ejemplo los enteros serían variables discretas, los reales continuas, los objetos categóricas, etc.

A un nivel más elevado podemos agruparlos en dos grupos que son super relevantes, ya que estos dos grupos van a determinar qué técnicas analíticas o qué gráficos podemos aplicar en cada uno de ellos:

* **Categóricas (o cualitativas)**: incluyen nominales y ordinales
* **Numéricas (o cuantitativas)**: incluyen discretas y razón

Como este curso tiene un caracter eminentemente aplicado vamos a usar esta división para ver qué tipo de análisis podemos hacer en cada una.

### Análisis de variables categóricas

##### Conteos y frecuencias

Es el valor más frecuente de una variable.

##### Tablas cruzadas

Consiste en cruzar dos o más variables para analizar la frecuencia.

##### Chi Cuadrado


#### Nota Codex - Cuándo usar chi-cuadrado

El test chi-cuadrado se usa para analizar si existe asociación entre **dos variables categóricas**. No mide fuerza lineal como Pearson; compara frecuencias observadas contra frecuencias esperadas si no hubiera relación.

Lectura rápida: si el `pvalor` es menor que `alpha`, rechazamos la independencia entre las variables.


Al hacer la tabla cruzada hemos visto que hay diferencias en si es fumador o no en base a si es hombre o mujer.

Pero, ¿estas diferencias son suficientes para tomar el resultado como una conclusión?. O quizá están dentro del margen esperable por efecto del muestreo y por tanto no podemos concluir nada.

Es lo que se llama ver si una conclusión es estadísticamente significativa. Y veremos el concepto en profunididad en la parte de estadística inferencial.

Pero en un análisis de tablas cruzadas como este podemos hacerlo con el estadístico chi-cuadrado.

Que parte de una hipótesis nula de que no hay relación entre las variables, es decir que son independientes y por tanto fumar o no, no depende del sexo, y la contrasta contra los datos obtenidos.

Si el pvalor es menor o igual que 0.05 entonces rechaza esa hipótesis nula, y significa que los datos si apoyan que hay diferencias significativas y por tanto podemos concluir que sí hay diferencias en cuanto al fumar entre hombres y mujeres.

````python
#Chi-cuadrado
tabla = pd.crosstab(df.sex,df.smoker)
chi, pvalor, gl, experado = sp.stats.chi2_contingency(tabla)
print("Chi cuadrado:", chi)
print("P-valor:", pvalor)
#En este caso el valor es mayor que 0.05, por tanto no podemos rechazar la hipótesis nula, 
#y por tanto no hay diferencias significativas entre hombres y mujeres
````

### Análisis de variables cuantitativas

#### Con cálculos

Se diferencian dos tipos de medida:

* De **centralización**: que intentan resumir la información de la variable en un sólo dato que pueda ser lo más representativo posible. Aquí están medidas como las medias, mediana o moda.

* De **dispersión**: que representan el grado de variabilidad existente en la variable. Normalmente cuanto más baja sea la medida de dispersión más representativa será la medida de centralización. Aquí entran métricas como la varianza, la desviación típica o el coeficiente de variación.

También veremos las métricas para cuantificar la relación existente entre dos variables cuantitativas: las correlaciones.

##### Medias


#### Nota Codex - Media o mediana

Usa la **media** cuando la variable no tiene valores extremos importantes y la distribución es razonablemente equilibrada. Usa la **mediana** cuando hay outliers o una distribución muy asimétrica, porque la mediana resiste mejor los valores extremos.


La media normal se llama **media aritmétrica**, y será la que usaremos en la mayoría de los casos.

La **media winsorizada** se usa cuando hay valores atípicos que pueden sesgar el valor de la media.

Para hacer la media de porcentajes o tasas hay que usar la **media geométrica**.

Para hacer la media de medias hay que usar la **media armónica**.

````python
#Media arimética
statistics.mean([3,4,5])
````

#### Nota Codex - Media winsorizada y valores extremos

En este ejemplo tenemos un array con varios valores pequeños y un valor extremadamente grande: `999997`. Ese valor extremo arrastra la media hacia arriba y hace que la media aritmética deje de representar bien al conjunto.

La **winsorización** no elimina el valor extremo: lo sustituye por un valor menos extremo dentro de los límites definidos. En este caso:

```python
limits=[0, 0.2]
```

significa:

- `0`: no se modifica nada por la parte baja de la distribución;
- `0.2`: se winsoriza el 20% superior de los datos.

Como hay 5 valores, el 20% equivale a 1 valor. Por eso el valor más alto, `999997`, se sustituye por el siguiente valor permitido, que es `6`.

Así podemos comparar:

- la media original, muy afectada por el valor extremo;
- la media winsorizada, más representativa de los valores habituales.


````python
#Media winsorizada
numeros = np.array([3,4,5,6,999997])
print(numeros)
print(numeros.mean())

#Winsorizar sustituye los valores fuera de los límites por el último valor
winsorizados = sp.stats.mstats.winsorize(numeros, limits = [0, 0.2])
print(winsorizados)
print(winsorizados.mean())
````

````python
#Comparamos con lo que hubiera salido en una media arimética
# tenemos un valor de media muy alto (200003) que no refleja la realidad de los datos, 
# ya que tenemos un valor atípico (999997) que está influyendo en la media. 
# Por eso es importante conocer la media winsorizada, ya que nos da una idea más realista de los datos.
statistics.mean([3,4,5,6,999997]) 
````

#### Nota Codex - Media geométrica

La **media geométrica** se usa cuando queremos promediar valores que representan crecimientos, tasas, porcentajes o factores multiplicativos.

La media aritmética suma y divide. La media geométrica, en cambio, multiplica los valores y después calcula una raíz. Por eso respeta mejor la lógica de los porcentajes encadenados.

En este ejemplo:

```python
statistics.geometric_mean([0.3, 0.4, 0.5])
```

estamos calculando una media adecuada para valores proporcionales. La idea es: si esos valores representan tasas o factores comparables, la media geométrica resume el crecimiento típico sin tratarlo como una simple suma.

Importante: la media geométrica solo tiene sentido con valores positivos. Si hay ceros o valores negativos, no se debe usar directamente.


````python
#Media geométrica (para porcentajes)
statistics.geometric_mean([0.3,0.4,0.5])
````

````python
#Comparamos con lo que hubiera salido en una media arimética
statistics.mean([0.3,0.4,0.5])
````

#### Nota Codex - Media armónica

La **media armónica** se usa para promediar razones, velocidades o tasas cuando todas las observaciones tienen el mismo peso de base.

Por ejemplo, si hablamos de velocidades medias en varias calles, no conviene hacer siempre una media aritmética simple, porque las velocidades son una relación entre distancia y tiempo. La media armónica corrige ese efecto dando más peso a los valores pequeños.

En este ejemplo:

```python
statistics.harmonic_mean([30.3, 40.8, 50.1])
```

calculamos una media más adecuada para magnitudes tipo "por unidad", como km/h, coste por unidad, ratios o rendimientos.

Intuición rápida: si una velocidad es baja, afecta mucho al promedio real del recorrido. Por eso la media armónica suele ser menor que la media aritmética.

Importante: la media armónica requiere valores positivos y no funciona bien si hay ceros.


````python
#Media armónica (para media de medias)
#Suponemos que los datos de esta lista son medias, ej velocidad media de coches en 3 calles
statistics.harmonic_mean([30.3,40.8,50.1])
````

````python
#Comparamos con lo que hubiera salido en una media arimética
statistics.mean([30.3,40.8,50.1])
````

##### Mediana

Si se ordenan todos los valores de la variable en orden ascendente o descendente la medianta es el valor de la variable correpondiente al elemento que ocupa la posición central, es decir, el que está en el 50%.

La mediana es una medida de centralización más recomendable que la media cuando tenemos distribuciones que no son normales, o cuando tenemos atípicos.

````python
#Mediana
# con el número de datos impar, la mediana es el valor central
numeros = np.random.randint(0,11,11)
numeros_ord = np.sort(numeros)
print(numeros_ord)
print(statistics.median(numeros_ord))
````

````python
#Mediana
# cuando el número de datos es par, la mediana es el promedio de los dos valores centrales
numeros = np.random.randint(0,11,10)
numeros_ord = np.sort(numeros)
print(numeros_ord)
print(statistics.median(numeros_ord))
````

##### Varianza


#### Nota Codex - Varianza frente a desviación típica

La varianza es útil matemáticamente, pero queda en unidades al cuadrado. Para interpretar en la escala original de los datos, normalmente usamos la desviación típica.


La varianza es el resultado de restar la media a cada valor de la variable, elevarlo al cuadrado (para evitar los negativos), sumarlo todo, y dividir el resultado por el número de datos.

Es una medida de dispersión, porque será mayor cuanto más lejos estén el global de los valores con respecto a la media.

````python
#Vamos a calcularla manualmente para entenderla
var1 = np.random.randint(0,11,20)
print(var1)
media = var1.mean()
print(media)
suma_cuadrados = sum((var1 - media) ** 2)
print(suma_cuadrados)
varianza = suma_cuadrados / (len(var1)) # len(var1) es el número de observaciones, en este caso 20
print(varianza)
````

````python
#En la práctica usaremos una función para calcularla directamente
var1.var()
````

##### Desviación típica

El problema con la varianza es que hemos tenido que elevar al cuadrado para quitarnos los negativos.

Entonces el dato obtenido ya no está en la misma escala que la media por ejemplo para poder compararlos.

Como solución se hace la raiz cuadrada a la varianza para volver a traerla a la escala, y es lo que se llama desviación típica.

````python
#Desviación típica
var1.std()
````

##### Coeficiente de variación

Como decíamos más arriba cuando más baja sea la dispersión con respecto a la media será mejor porque nos indica que esa media es más representativa del total de los datos de la variable.

Ahora que ya tenemos una medida, la desviación típica, que sí está en la misma escala que la media, podemos compararlas.

Es lo que se llama el coeficiente de variación, y consiste en dividir la desviación típica por la media. Normalmente se pone en porcentaje.

Permite comparar entre variables de diferentes escalas, por ejemplo: ¿en qué somos más variables las personas: en la altura o en el peso?

#### Nota Codex - Por qué comparamos altura y peso

Hasta aquí usamos el coeficiente de variación para analizar una variable: comparamos su desviación típica con su propia media para saber si la media representa bien esos datos.

Pero también sirve para comparar la variabilidad de **variables diferentes**, incluso si tienen unidades distintas.

Por ejemplo, no podemos comparar directamente:

- desviación típica de la altura en centímetros;
- desviación típica del peso en kilos.

`10 cm` y `10 kg` no son comparables directamente. Por eso usamos la variabilidad relativa:

```python
CV altura = desviación_altura / media_altura
CV peso = desviación_peso / media_peso
```

Ejemplo:

```text
CV altura = 10 / 170 = 5.8%
CV peso = 10 / 70 = 14.3%
```

En este caso, el peso es más variable que la altura **en relación con su propia media**.



````python
#Coeficiente de variación
var1.std() / var1.mean() * 100
````

##### Correlación de Pearson


#### Nota Codex - Mini guía para interpretar correlaciones

Una regla práctica orientativa:

- Cerca de `0`: relación lineal débil o inexistente.
- Entre `0.1` y `0.3`: relación baja.
- Entre `0.3` y `0.6`: relación moderada.
- Por encima de `0.6`: relación fuerte.

El signo indica dirección: positivo si suben juntas, negativo si una sube cuando la otra baja.


Cuando la gente habla de correlación para medir la relación entre dos variables se está refiriendo a Pearson.

Sólo sería técnicamente correcto usar Pearson con variables contínuas y linealmente relacionadas, y si estas tienen una distribución normal.

Pero si las variables no son contínuas o normalmente distribuídas, o su relación es no lineal usar Pearson no es lo más correcto. Podríamos usar Spearman como veremos más abajo.

Además la correlación de Pearson es una medida lineal, y por tanto no válida para medir relaciones no lineales.

Ello significa que si dos variables aparecen como correlacionadas según Pearson efectivamente tendrán relación. Pero lo contrario no es siempre cierto, ya que podría haber relación pero ser no lineal.

La forma más fácil para identificar correlaciones no lineales es usar un gráfico de dispersión como veremos más adelante.

La correlación de Pearson se interpreta así:

* +1 o -1: es una correlación perfecta (siempre que sube una la otra también o siempre que sube una la otra baja)
* 0: no hay relación entre las variables
* Si el pvalor está por debajo de 0.05 la relación sí es significativa (se recomienda considerarlo sólo para más de 500 datos)

````python
#Correlación de Pearson
#Esta implementación de scipy devuelve la correlación como primer valor y el pvalor como el segundo
var1 = np.random.randint(1,21,1000)
var2 = np.random.randint(1,21,1000)

sp.stats.pearsonr(var1,var2)
````

##### R cuadrado

R cuadrado es simplemente el cuadrado de la correlación de Pearson.

Pero sin embargo es una métrica muy útil en modelización, ya que se puede entender como el porcentaje de una variable que podemos explicar a partir de otra (o combinación de otras como en los modelos lineales).

Por tanto será frecuente verla en modelos como las regresiones, donde el software nos reportará por ejemplo un R cuadrado de 0.6 y eso significa que con nuestro modelo estamos siendo capaces de explicar el 60% de la variable objetivo, y por tanto nos queda un 40% que no sabemos explicar.

Al igual que la correlación de Pearson es una medida lineal, por tanto si cuando estamos modelizando con una técnica lineal nos sale un R cuadrado bajo puede ser interesante probar una técnica no lineal para ver si hay relaciones no lineales que no estaban siendo capturadas.

#### Nota Codex - Pearson, R² y Spearman: por qué existen los tres

Los tres conceptos están relacionados, pero no responden exactamente a la misma pregunta.

**Pearson** mide la fuerza y dirección de una relación **lineal** entre dos variables numéricas. Úsalo cuando quieres saber si dos variables suben o bajan juntas siguiendo una relación aproximadamente recta.

**R²** es el cuadrado de Pearson cuando hablamos de dos variables. Se interpreta como proporción explicada: por ejemplo, `R² = 0.64` significa que aproximadamente el 64% de la variabilidad de una variable se explica linealmente por la otra. Es especialmente útil en modelos de regresión.

**Spearman** mide si existe una relación **monótona**, es decir, si cuando una variable aumenta la otra tiende a aumentar o disminuir, aunque la relación no sea una línea recta. Trabaja con rangos, por eso es más robusto ante outliers y sirve mejor para variables ordinales.

Regla práctica:

- usa **Pearson** si buscas relación lineal entre variables numéricas;
- usa **R²** si quieres interpretar cuánta variabilidad queda explicada por una relación lineal o un modelo;
- usa **Spearman** si la relación parece ordenada pero no necesariamente lineal, o si hay outliers / variables ordinales.


##### Correlación de Spearman

La mayoría de la gente conoce la correlación de Pearson para medir la relación entre dos variables.

Pero realmente sólo es correcto usar Pearson con variables contínuas, linealmente relacionadas y si estas tienen una distribución normal.

Pero si las variables no son contínuas o no son normalmente distribuídas usar Pearson no es lo más correcto.

La **Rho de Spearman** proporciona una solución para hacer la correlación cuando:

* las variables son contínuas pero no se distribuyen según la normal
* las variables son rankings
* la relación es no lineal
* las variables son discretas pero con menos de 30 valores distintos

Se interpreta así:

* +1 o -1: es una relación perfectamente monotónica (siempre que sube una la otra también o al revés)
* 0: no hay relación entre las variables
* Si el pvalor está por debajo de 0.05% la relación sí es significativa (se recomienda considerarlo sólo para más de 500 datos)

##### Correlación Vs Causalidad


#### Nota Codex - Correlación no implica causa

Una correlación puede aparecer por causalidad real, casualidad, una tercera variable oculta o una relación indirecta. Para hablar de causa necesitamos diseño experimental, control de variables o métodos causales específicos.


Este es uno de los errores más frecuentes en el uso cotidiano de la estadística.

Se tiende a pensar que si dos variables están correlacionadas entonces se puede establecer una relación causa - efecto entre ellas.

Realmente, para poder establacer una relación causa - efecto se tiene que cumplir que:

1. Exista correlación
2. La causa preceda en el tiempo a la consecuencia
3. Se puedan descartar explicaciones alternativas

Aquí entran los conceptos de correlación espúrea y correlación parcial.

**Correlación espúrea** es cuando dos variables parece que están relacionadas (pueden correlacionar de hecho), pero es realmente por el efecto de otras terceras variables no consideradas.

**Correlación parcial** es la correlación real que queda entre las 2 primeras variables una vez que se elimina el efecto de la tercera o terceras.

Por ejemplo, está demostrado que existe alta correlación entre el tamaño del pie y el cociente de inteligencia.

Sin embargo esa correlación está marcada realmente por una tercera variable, la edad. Si se controla la edad y se elimina su efecto entonces ya no existe correlación entre el tamaño del pie y el cociente de inteligencia.

En contextos empresariales está plagado de este tipo de efectos y es conveniente conocer este concepto y aplicar siempre la visión crítica antes de obtener conclusiones.

Ejemplos:

- No suben las ventas cuando hay promociones: ¿se han controlado las promociones de la competencia?
- Los empleados que viajan dejan más la empresa: ¿se ha controlado la edad?
- Los clientes que más ganan sin embargo tienen menos ahorros: ¿se ha diferenciado entre los que tienen hipoteca y los que no?

### Crea dos vectores de enteros aleatorios entre 1 y 20 y calcula la correlación de Spearman, indicando si es estadísticamente significativa

````python
import scipy as sp

x = np.random.randint(1, 20, 1000)
y = np.random.randint(1, 20, 1000)

corr, p_valor = sp.stats.spearmanr(x, y)

print("correlación de Spearman:", corr)
print("valor p:", p_valor)

# Interpretación del sentido de la relación
if corr > 0:
    print("La correlación es positiva: cuando x aumenta, y tiende a aumentar.")
elif corr < 0:
    print("La correlación es negativa: cuando x aumenta, y tiende a disminuir.")
else:
    print("La correlación es exactamente 0: no hay relación monótona.")

# Interpretación de la fuerza de la relación
corr_abs = abs(corr)

if corr_abs < 0.1:
    print("La fuerza de la correlación es muy débil o casi nula.")
elif corr_abs < 0.3:
    print("La fuerza de la correlación es débil.")
elif corr_abs < 0.6:
    print("La fuerza de la correlación es moderada.")
else:
    print("La fuerza de la correlación es fuerte.")

# Interpretación de la significación estadística
if p_valor < 0.05:
    print("La correlación es estadísticamente significativa.")
else:
    print("La correlación no es estadísticamente significativa.")

# Ojo, con numeros aleatorios, la correlación puede ser significativa pero no tener sentido práctico. 
# Siempre hay que interpretar los resultados en el contexto del problema que se está estudiando.
````

### Calcula el R cuadrado entre dos vectores de enteros aleatorios generados entre 1 y 20

````python
x = np.random.randint(1, 21, 1000)
y = np.random.randint(1, 21, 1000)

# Como son números aleatorios muy probablemente no tengan correlación,
# pero vamos a calcular el R cuadrado para ver si hay alguna relación lineal entre las dos variables.
pearson_corr, p_valor = sp.stats.pearsonr(x, y)

r2 = pearson_corr ** 2
r2_pct = r2 * 100

print("correlación de Pearson:", pearson_corr)
print("R cuadrado:", r2_pct, "%")
print("valor p:", p_valor)

if r2_pct < 1:
    print("Interpretación: capacidad explicativa prácticamente nula. Una variable no explica linealmente a la otra.")
elif r2_pct < 5:
    print("Interpretación: capacidad explicativa muy baja. La relación lineal existe como mucho de forma muy débil.")
elif r2_pct < 15:
    print("Interpretación: capacidad explicativa baja. Hay algo de relación lineal, pero no es suficiente para explicar bien la variable.")
elif r2_pct < 40:
    print("Interpretación: capacidad explicativa moderada. La relación lineal puede ser útil, pero todavía deja mucha variabilidad sin explicar.")
elif r2_pct < 70:
    print("Interpretación: capacidad explicativa alta. La relación lineal explica una parte importante de la variabilidad.")
else:
    print("Interpretación: capacidad explicativa muy alta. Una variable explica linealmente gran parte de la variabilidad de la otra.")

if p_valor < 0.05:
    print("Significación: la correlación lineal es estadísticamente significativa.")
else:
    print("Significación: la correlación lineal no es estadísticamente significativa.")
````

### Realiza un test chi-cuadrado entre las variables sex y day del dataset tips e imprime el valor p del test

````python
tabla = pd.crosstab(df.sex, df.day)

chi2, pvalor, gl, esperados = sp.stats.chi2_contingency(tabla)
# gl = grados de libertad = (número de filas - 1) * (número de columnas - 1)
# esperados = tabla.values.sum() * (tabla.sum(axis=1).values / tabla.values.sum())[:, np.newaxis] * (tabla.sum(axis=0).values / tabla.values.sum())[np.newaxis, :]

print("tabla observada:")
print(tabla)

print("\nchi-cuadrado:", chi2)
print("grados de libertad:", gl)
print("p-valor:", pvalor)

alpha = 0.05

if pvalor < alpha:
    print("\nInterpretación:")
    print("Rechazamos la hipótesis nula.")
    print("Hay evidencia estadística de que sex y day NO son independientes.")
    print("Es decir, la distribución de sex parece cambiar según el día.")
else:
    print("\nInterpretación:")
    print("No rechazamos la hipótesis nula.")
    print("No hay evidencia estadística suficiente para decir que sex y day estén relacionados.")
    print("Es decir, la distribución de sex parece similar entre los distintos días.")
````

## 1. Cuartiles, percentiles e IQR

#### Nota Codex

Los percentiles dividen los datos según posición. El percentil 25 deja por debajo el 25% de los datos. Los cuartiles más usados son Q1, Q2 y Q3. El IQR, o rango intercuartílico, se calcula como `Q3 - Q1` y mide la dispersión central sin dejarse dominar por extremos.


````python

# Ejemplo Codex: cuartiles e IQR
import numpy as np

datos = np.array([12, 14, 15, 18, 19, 21, 22, 23, 24, 100])
q1 = np.percentile(datos, 25)
q2 = np.percentile(datos, 50)
q3 = np.percentile(datos, 75)
iqr = q3 - q1

print('Q1:', q1)
print('Q2 / mediana:', q2)
print('Q3:', q3)
print('IQR:', iqr)
````


## 2. Detección de outliers con IQR

#### Nota Codex

Una regla clásica considera outliers los valores menores que `Q1 - 1.5 * IQR` o mayores que `Q3 + 1.5 * IQR`. No es una ley universal, pero es un buen primer filtro exploratorio.


````python

# Ejemplo Codex: outliers con IQR
# IQR = rango intercuartílico, que es la diferencia entre el tercer cuartil (Q3) y el primer cuartil (Q1). 
# Se utiliza para identificar valores atípicos (outliers) en un conjunto de datos. 
# Los límites inferior y superior se calculan restando y sumando 1.5 veces el IQR a Q1 y Q3, respectivamente. 
# Cualquier valor fuera de estos límites se considera un outlier.
limite_inferior = q1 - 1.5 * iqr 
limite_superior = q3 + 1.5 * iqr
outliers = datos[(datos < limite_inferior) | (datos > limite_superior)]

print('Límite inferior:', limite_inferior)
print('Límite superior:', limite_superior)
print('Outliers:', outliers)
````


## 3. Asimetría y curtosis

#### Nota Codex

La **asimetría** indica si la distribución tiene una cola más larga hacia un lado. La **curtosis** ayuda a describir el peso de las colas. Son útiles para detectar distribuciones alejadas de la normalidad.

# Le Curtosis (ou Kurtosis)

Le **curtosis** est une mesure statistique qui indique le degré d'aplatissement et l'épaisseur des queues d'une distribution de données par rapport à une distribution normale. Il permet de détecter la présence de valeurs extrêmes ou aberrantes (*outliers*).

---

## Les Trois Types de Formes

* **Mésokurtique :** Distribution identique à la loi normale (valeur de référence = 3 en brut, ou 0 en excès).
* **Leptokurtique :** Courbe pointue avec des queues épaisses (valeur supérieure à la normale). Risque élevé de valeurs extrêmes.
* **Platykurtique :** Courbe aplatie avec des queues fines (valeur inférieure à la normale). Faible présence de valeurs extrêmes.



````python

# Ejemplo Codex: skewness y kurtosis
from scipy import stats

print('Asimetría:', stats.skew(datos))
print('Curtosis:', stats.kurtosis(datos))
````

````python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# 1. Cálculo de las métricas
asimetria = stats.skew(datos)
curtosis_exceso = stats.kurtosis(datos)  # Por defecto, scipy calcula la curtosis en exceso (normal = 0)

print(f"Asimetría: {asimetria:.4f}")
print(f"Curtosis (en exceso): {curtosis_exceso:.4f}")
print("-" * 40)

# 2. Interpretación automática de la curtosis
if curtosis_exceso > 0.5:
    print("Interpretación: distribución leptocúrtica (curva más apuntada, colas pesadas, posible presencia de outliers).")
elif curtosis_exceso < -0.5:
    print("Interpretación: distribución platicúrtica (curva más plana, colas ligeras, pocos outliers).")
else:
    print("Interpretación: distribución mesocúrtica (cercana a una distribución normal).")

# 3. Visualización gráfica
plt.figure(figsize=(8, 5))
plt.hist(datos, bins=30, density=True, alpha=0.6, color='g', edgecolor='black', label='Datos')

# Añadimos una curva normal teórica para comparar
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = stats.norm.pdf(x, np.mean(datos), np.std(datos))
plt.plot(x, p, 'r--', linewidth=2, label='Normal teórica')

plt.title('Visualización de la distribución (asimetría y curtosis)')
plt.xlabel('Valores')
plt.ylabel('Densidad')
plt.legend()
plt.show()
````
